# Easy Level (1–10)

#### 1. Write a lambda function to return the square of a number.

```python
square = lambda x: (x*x)
square(5)
```

#### 2. Write a lambda function to return the cube of a number.

```python
cube = lambda x:(x**3)
cube(4)
```

#### 3.Write a lambda function to add two numbers.

```python
add = lambda a,b: (a+b)
add(10,20)
```

#### 4. Write a lambda function to find the maximum of two numbers.

```python
Max = lambda a,b: max(a,b)
Max(15,8)
```

#### 5.Write a lambda function that returns "Even" if a number is even; otherwise "Odd".

```python
Even_Odd = lambda num: "Even"if num % 2 == 0 else "Odd"
Even_Odd(2)
```

#### 6. Use map() with lambda to square every element of a list.

```python
numbers = [ 2,3,4,5,6]

Map = list(map(lambda x : x**2,numbers))

Map
```

#### 7. Use filter() with lambda to keep only even numbers.

```python
number = [1,2,3,4,5,6,7]

Filter = list(filter(lambda x : x % 2 == 0, number))

Filter
```

#### 8. Use filter() with lambda to keep numbers greater than 50.

#### 9.Sort a list of numbers in ascending order using lambda.

```python
number = [ 6,3,2,5,4,1]

Sort = lambda x : sorted(x)

Sort(number)
```

#### 10.Sort a list of numbers in descending order using lambda.




```python
number = [1,2,3,4,5,6,7,8,9]

sort = lambda x : sorted(x,reverse=True)

sort(number)
```

#### 11. Use reduce() with lambda to find the sum of all numbers.

```python
from functools import reduce

number = [1,2,3,4,5,6,7,8,9]

result =  reduce(lambda x,y: x+y, number)

print(result)
```

#### 12. Use reduce() with lambda to find the product of all numbers.

```python
from functools import reduce

number = [2,3,4,5]

result =  reduce(lambda x,y: x*y, number)

print(result)
```

#### 21. Sort the following list of dictionaries based on salary.

```python
employees = [
{"name":"Ram","salary":50000},
{"name":"Amit","salary":65000},
{"name":"Neha","salary":45000}
]

sorted_employees = sorted(employees, key = lambda emp: emp["salary"])

print(sorted_employees)
```

#### 22. Sort students by marks descending and name ascending.

```python
students = [
("Ram",90),
("Amit",90),
("Neha",85),
("Raj",95)
]

descending_ascending = sorted(students, key = lambda x: (-x[1],x[0]))

print(descending_ascending)
```

```python
#### 23. Use nested lambda to multiply two numbers.

Multiply = lambda x: (lambda y: x*y)

result = Multiply(5)(9)

print(result)
```

#### 24. Create a lambda that returns another lambda for exponentiation.

```python
power = lambda x: (lambda y: x ** y)

result = power(2)(3)

print("Result:", result)
```

#### 25. Using lambda and filter(), remove all negative numbers.

```python
number = [-5,10,-3,8,0,-1,20]

Remove_Nigative = list(filter(lambda x: x > 0,number))

print(Remove_Nigative)
```

#### 26.Using lambda and map(), replace every odd number by its square while leaving even numbers unchanged.

```python
num = [1,2,3,4,5,6]

result = list(map(lambda x: x**2 if x % 2 != 0 else x, num))

print("Original List:", num)
print("Modified List:", result)
```
