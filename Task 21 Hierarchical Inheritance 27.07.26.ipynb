# Basic Hierarchical Inheritance (1–12)

#### 1. Person → Student, Teacher
#### Question
#### Create a parent class Person with attributes name and age. Create two child classes Student
#### and Teacher that inherit from Person.
#### Concept
#### • Hierarchical Inheritance
#### • Parent attributes inherited by multiple child classes.
#### How to Solve
#### 1. Create a parent class Person.
#### 2. Add name and age using a constructor.
#### 3. Create Student(Person).
#### 4. Create Teacher(Person).
#### 5. Create objects of both classes.
#### 6. Show that both children can access name and age.

```python
# Parent class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age


# Child class 1
class Student(Person):
    def display_student(self):
        print("Student Name:", self.name)
        print("Student Age:", self.age)


# Child class 2
class Teacher(Person):
    def display_teacher(self):
        print("Teacher Name:", self.name)
        print("Teacher Age:", self.age)


# Creating objects
s1 = Student("Rohit", 22)
t1 = Teacher("Yash", 21)

# Display details
s1.display_student()
print()
t1.display_teacher()
```

#### 2. Animal → Dog, Cat
#### Question

#### How to Solve
#### 1. Create parent class Animal.
#### 2. Write method eat().
#### 3. Create child classes Dog and Cat.
#### 4. Create objects.
#### 5. Call eat() using both objects.

#### 3. Vehicle → Car, Bike
#### Question
#### Create class Vehicle with method start(). Create subclasses Car and Bike.

#### How to Solve
#### 1. Create Vehicle.
#### 2. Add start().
#### 3. Inherit it in Car.
#### 4. Inherit it in Bike.

#### 5. Employee → Manager, Developer
#### Question
#### Create class Employee with attributes emp_id and salary. Create subclasses Manager and
#### Developer.
#### Concept
#### • Attribute inheritance.
#### How to Solve
#### 1. Create parent constructor.
#### 2. Store emp_id and salary.
#### 3. Create two child classes.
#### 4. Show inherited attributes.

```python
# Parent class
class Employee:
    def __init__(self, emp_id, salary):
        self.emp_id = emp_id
        self.salary = salary


# Child class 1
class Manager(Employee):
    def __init__(self, emp_id, salary):
        super().__init__(emp_id, salary)

    def display(self):
        print("Manager Details")
        print("Employee ID:", self.emp_id)
        print("Employee Salary:", self.salary)


# Child class 2
class Developer(Employee):
    def __init__(self, emp_id, salary):
        super().__init__(emp_id, salary)

    def display(self):
        print("Developer Details")
        print("Employee ID:", self.emp_id)
        print("Employee Salary:", self.salary)


# Creating objects
manager = Manager("SUMIT123", 100000)
developer = Developer("RD234", 20000)

# Display details
manager.display()
print()
developer.display()
```

#### 10. Book → Fiction, NonFiction
#### Question
#### Define class Book with attributes title and author.
#### Concept
#### • Constructor inheritance.
#### How to Solve
#### 1. Parent constructor.
#### 2. Two child classes.
#### 3. Access inherited attributes.

```python
class Book:
    def __init__(self,title,author):
        self.Book_Title = title
        self.Book_Author = author


class Fiction(Book):
    def display(self):
        print("Fiction Book")
        print("Book Title:", self.Book_Title)
        print("Book Author::",self.Book_Author)

class NonFiction(Book):
    def display(self):
        print("Non-Fiction Book")
        print("Book Title:", self.Book_Title)
        print("Book Author::",self.Book_Author)


T = Fiction("Rich Dad Poor Dad","Robert T. Kiyosaki")
A = NonFiction("The Hobbit", "J.R.R. Tolkien")


T.display()
print()
A.display()
```

#### 11. Account → UserAccount, AdminAccount
#### Question
#### Define class Account with method show_balance().
#### Concept
#### • Shared functionality.
#### How to Solve
#### 1. Create method.
#### 2. Inherit into both children.

```python
class Account:
    def __init__(self, balance):
        self.balance = balance

    def show_balance(self):
        print("Balance:", self.balance)


class User_Account(Account):
    pass

class AdminAccount(Account):
    pass

usr = User_Account(5000)
adm = AdminAccount(10000)

print("User Account")
usr.show_balance()
print("-------------")
print("Admin Account")
adm.show_balance()
```

# Hierarchical Inheritance with Constructors & Methods (13–22)

#### 13. Person Constructor
##### Concept
##### Using super() to call the parent constructor.
###### How to Solve
##### 1. Create constructor in parent.
##### 2. Create child constructors.
##### 3. Use super().__init__().
##### 4. Add child-specific attribute.
##### 5. Display all data.

```python
# Parent class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Child class
class Student(Person):
    def __init__(self, name, age, roll_no):
        super().__init__(name, age)
        self.roll_no = roll_no

    def display(self):
        print("Student Details")
        print("Name    :", self.name)
        print("Age     :", self.age)
        print("Roll No :", self.roll_no)

# Child class
class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject = subject

    def display(self):
        print("Teacher Details")
        print("Name    :", self.name)
        print("Age     :", self.age)
        print("Subject :", self.subject)

# Creating objects
s = Student("Rohit", 22, 101)
t = Teacher("Yash", 21, "Mathematics")

# Display details
s.display()
print()
t.display()
```
