### Basic Exploration & Missing Values

#### 1. How many missing values are there in each column? Which column has the highest number of missing values?

```python
import pandas as pd

df = pd.read_csv("C:\\Users\\rohit\\Downloads\\Customer_datasets.csv")

a = df.isna().sum()
print("Missing Values",a)

Highest_Missing = a.max()
Highest_column = a.idxmax()
print("\nHighest Missing Value Column",Highest_Missing)
print("\nHighest Column",Highest_column)
```

#### 2. What percentage of rows have at least one missing value?

```python
percentage_missing = df.isnull().any(axis=1).mean() * 100

percentage_missing
```

#### 3. Drop all rows that have more than 3 missing values. How many rows remain?

```python
df_cleaned = df[df.isnull().sum(axis=1) <= 3]

df_cleaned
```

## Age Column

#### 6. Convert the Age column to numeric. How many values could not be converted (became NaN)?

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

invalid_age = df["Age"].isna().sum()

invalid_age
```

#### 7. After converting Age to numeric, identify and remove outliers using the IQR method. How many outliers were removed?

```python
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = (df["Age"] < lower_bound) | (df["Age"] > upper_bound)

num_outliers = outliers.sum()

df = df[~outliers]

print("Number of outliers removed:", num_outliers)
```

#### 8. Replace remaining missing values in Age with the median age.

```python
df["Age"] = df["Age"].fillna(df["Age"].median())

df["Age"].isna().sum()
```

#### 9. Clean the Salary column: remove currency symbols ($), commas, and convert
#### everything possible to numeric. How many values remained non-numeric?

```python
df["Salary"] = (
    df["Salary"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")

non_numeric_salary = df["Salary"].isna().sum()

non_numeric_salary
```

#### 10. After cleaning, find the mean and median salary. Which one is more affected by outliers?

```python
mean_salary = df["Salary"].mean()
median_salary = df["Salary"].median()

print("Mean Salary:", mean_salary)
print("Median Salary:", median_salary)
```
