# Problem-16
### Write a Python program to calculate the area of a circle.
### Use the formula Area = π × r².
### Store the value of π as 3.14 in a variable.
### Accept the radius as input and display the area. use round to round the digits upto 2 decimal values

```python
pi = 3.14
r = float(input())

area = pi * r ** 2

print(round(area, 2))
```

# Problem-17
### Write a Python program to calculate the circumference of a circle.
### Use the formula Circumference = 2 × π × r.
### Store π as 3.14 in a variable.
### Accept radius as input and print the result. use round to get result upto 2 decimal values


```python
pi = 3.14
r = float(input())

circumference = 2 * pi * r

print(round(circumference, 2))
```
