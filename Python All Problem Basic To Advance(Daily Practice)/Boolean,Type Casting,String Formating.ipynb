# Problem-4
### Write a Python program to check the strength of a password.
### A password is strong if it has at least 8 characters, one uppercase letter, one digit, and one special character.
#### Use conditions to verify all these requirements.
### Print "Strong Password" if all conditions are satisfied, otherwise print "Weak Password".


```python

password = input()

if (len(password) >= 8 and
    any(ch.isupper() for ch in password) and
    any(ch.isdigit() for ch in password) and
    any(not ch.isalnum() for ch in password)):
    print("Strong Password")
else:
    print("Weak Password")
```

### Take an integer (whole number) from the user
### Take a float (decimal number) from the user

### Store the result in a variable

# Problem-8
### In Python, when you receive input from a user, it always comes as a string (text), even if it looks like a number. For example, when you type 123, Python sees it as "123" (text), not the number 123.
### To perform calculations or use these values properly, we need to convert them to the correct data type. This process is called type conversion or type casting.
### What you need to do:

### Take a numeric string from the user (a number typed as text)
### Convert this string into three different types:

### Integer (int): Whole number
### Float: Decimal number
### Boolean (bool): True or False value


### Store each converted value in a separate variable
### Print each converted value along with its data type

### Key Learning:

### int() converts string to whole number
### float() converts string to decimal number
### bool() on the integer value shows True for non-zero numbers and False for zero
### The same string can be converted to multiple types depending on what you need

### Input Format:
### Single line containing a numeric string (a number in text form).
### Output Format:
### Six lines showing:

### Line 1: The integer value
### Line 2: The type of integer (will show <class 'int'>)
### Line 3: The float value
### Line 4: The type of float (will show <class 'float'>)

# Problem-9
### When working with numbers in Python, you often need to convert between different data types depending on your needs. A float (decimal number) can be transformed into other types, and each conversion behaves differently.
### What you need to do:

### Take a float value (decimal number) from the user
### Convert this float into three different types:

### Integer (int): This will remove the decimal part (truncation, not rounding)
### String (str): This will convert the number into text format
### Boolean (bool): This will show True or False based on the number's value


### Store each converted value in a separate variable
### Print each converted value along with its data type

### Key Learning:

### int() on a float truncates (cuts off) the decimal part. For example, 45.99 becomes 45, not 46
### str() converts the float to text, preserving the decimal representation
### bool() on a number returns False only when the number is exactly 0, otherwise it returns True
### We convert the integer value (not the string) to boolean to see the proper True/False behavior

### Input Format:
###  Single line containing a float value (decimal number).
### Output Format:
### Six lines showing:

### Line 1: The integer value (decimal part removed)
### Line 2: The type of integer (will show <class 'int'>)
### Line 3: The string value (number as text)
### Line 4: The type of string (will show <class 'str'>)
### Line 5: The boolean value (True or False)
### Line 6: The type of boolean (will show <class 'bool'>)

```python

num = float(input())

int_value = int(num)
str_value = str(num)
bool_value = bool(num)

print(int_value)
print(type(int_value))

print(str_value)
print(type(str_value))

print(bool_value)
print(type(bool_value))
```
