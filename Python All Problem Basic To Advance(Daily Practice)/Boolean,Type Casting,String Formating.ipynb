# Problem-4
### Write a Python program to check the strength of a password.
### A password is strong if it has at least 8 characters, one uppercase letter, one digit, and one special character.
#### Use conditions to verify all these requirements.
### Print "Strong Password" if all conditions are satisfied, otherwise print "Weak Password".


```python

password = input()

if (len(password) >= 8 and
    any(ch.isupper() for ch in password) and
    any(ch.isdigit() for ch in password) and
    any(not ch.isalnum() for ch in password)):
    print("Strong Password")
else:
    print("Weak Password")
```

# Problem-7
### In Python, numbers can be of different types. The two most common types are:

### Integer (int): Whole numbers without decimals, like 5, 10, -3
### Float: Numbers with decimal points, like 5.5, 10.0, -3.14

### When you perform operations between different number types, Python automatically handles the conversion. Understanding how this works is important for working with calculations.
### What you need to do:

### Take an integer (whole number) from the user
### Take a float (decimal number) from the user
### Add both numbers together using the + operator
### Store the result in a variable
### Print the sum
### Print the data type of the result using the type() function

### Key Learning:
### When you add an integer and a float, Python automatically converts the result to a float to preserve the decimal precision.
### Input Format:
### Two lines of input:

### First line: An integer (whole number)
### Second line: A float (decimal number)

### Output Format:
### Two lines:

### First line: The sum of both numbers
### Second line: The data type of the result (it will show <class 'float'>)

```python

num1 = int(input())
num2 = float(input())

result = num1 + num2

print(result)
print(type(result))
```
