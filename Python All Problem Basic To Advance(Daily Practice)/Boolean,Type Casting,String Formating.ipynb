# Problem-1
### Write a Python program to read a number from the user.
### Use an if condition to check whether the number is greater than 10.
### Print a message indicating whether the number is greater than 10 or not.

```python

num = int(input())

if num > 10:
    print("Greater than 10")
else:
    print("Not greater than 10")
```

# Problem-2
### Write a Python program to read a number from the user.
### Check whether the number is divisible by both 3 and 5.
### Use the modulo operator (%) to perform the check.
### Print a message indicating if the number is divisible by both or not.


```python

num = int(input())

if num % 3 == 0 and num % 5 == 0:
    print("Divisible by both 3 and 5")
else:
    print("Not divisible by both 3 and 5")
```

# Problem-4
### Write a Python program to check the strength of a password.
### A password is strong if it has at least 8 characters, one uppercase letter, one digit, and one special character.
#### Use conditions to verify all these requirements.
### Print "Strong Password" if all conditions are satisfied, otherwise print "Weak Password".


```python

password = input()

if (len(password) >= 8 and
    any(ch.isupper() for ch in password) and
    any(ch.isdigit() for ch in password) and
    any(not ch.isalnum() for ch in password)):
    print("Strong Password")
else:
    print("Weak Password")
```
