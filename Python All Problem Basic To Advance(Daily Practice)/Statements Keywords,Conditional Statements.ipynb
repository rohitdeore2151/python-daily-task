# Problem-8
### Write a Python program to accept a year as input.
### Check if the year is a leap year according to leap year rules.
### Print Leap Year or Not a Leap Year.


```python
year = int(input())

if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

# problem-9
### Write a Python program to accept marks of five subjects from the user.
### Calculate the average of the marks.
### Display the average and the corresponding grade.

### Grade Assignments
### A : Average greather than or equals to 90
### B : Average greather than or equals to 80
### C : Average greather than or equals to 70
### D : Average greather than or equals to 60
### E : Average greather than or equals to 50
### F : Average less than 50

```python
m1 = float(input())
m2 = float(input())
m3 = float(input())
m4 = float(input())
m5 = float(input())

average = (m1 + m2 + m3 + m4 + m5) / 5

if average >= 90:
    grade = "A"
elif average >= 80:
    grade = "B"
elif average >= 70:
    grade = "C"
elif average >= 60:
    grade = "D"
elif average >= 50:
    grade = "E"
else:
    grade = "F"

print("Average:", average)
print("Grade:", grade)
```
