# problem-9
### Write a Python program to accept marks of five subjects from the user.
### Calculate the average of the marks.
### Display the average and the corresponding grade.

### Grade Assignments
### A : Average greather than or equals to 90
### B : Average greather than or equals to 80
### C : Average greather than or equals to 70
### D : Average greather than or equals to 60
### E : Average greather than or equals to 50
### F : Average less than 50

```python
m1 = float(input())
m2 = float(input())
m3 = float(input())
m4 = float(input())
m5 = float(input())

average = (m1 + m2 + m3 + m4 + m5) / 5

if average >= 90:
    grade = "A"
elif average >= 80:
    grade = "B"
elif average >= 70:
    grade = "C"
elif average >= 60:
    grade = "D"
elif average >= 50:
    grade = "E"
else:
    grade = "F"

print("Average:", average)
print("Grade:", grade)
```


# Problem-12
### Write a Python program to convert a given USD amount to another currency based on the destination country.
### An integer: usd and a string: country
### Converted amount with currency label or "Unknown Country"
### (Use exchange rates like: India → INR 83, Japan → JPY 155, UK → GBP 0.78)

```python
usd, country = input().split()
usd = int(usd)

if country == "India":
    print(usd * 83, "INR")
elif country == "Japan":
    print(usd * 155, "JPY")
elif country == "UK":
    print(usd * 0.78,"GBP")
else:
    print("Unknown Country")
```

# Problem -18
### You need to write a Python program that works with two variables
### Create a variable named 'a' and assign it the value 10
### Create a variable named 'b' and assign it the value 5
### Your task is to swap (exchange) the values between these two variables without using any third variable
### After swapping, variable 'a' should contain 5 and variable 'b' should contain 10
### Display both variables before and after swapping to show the exchange happened
### This teaches you how to swap values efficiently using Python's built-in feature

```python
a = 10
b = 5

print("Before swapping:")
print("a =", a)
print("b =", b)

a, b = b, a

print("After swapping:")
print("a =", a)
print("b =", b)
```

# Problem-21
### Before we dive into the exercise, let's understand what reserved keywords actually are. Think of Python as a language with its own special words that have specific jobs. These words are called "reserved keywords" or "keywords" for short. They're reserved because you can't use them for your own variable names or function names - Python has already claimed them for special purposes.
### For example, words like if, else, for, and while are keywords because Python uses them to control how your program runs. If Python let you create a variable called if, it would get confused about whether you mean the word "if" for decision-making or your variable named "if". To avoid this confusion, these words are off-limits for regular use.
### Python has a built-in module called keyword that keeps track of all these reserved words. This module has a handy list that shows you every keyword currently in the Python language. As Python evolves, keywords can be added or removed, so using this module ensures you always get the current list.
### Now, let's work on a program that displays all these reserved keywords so you can see exactly which words are special in Python.

### Write a Python program that shows all the reserved keywords in the Python language
### You'll need to use Python's built-in keyword module to access this list
### The program should import the keyword module and then display all keywords
### No user input is needed - the keywords are already stored inside Python
### This helps you understand which words you cannot use as variable names in your programs


```python
import keyword

print(keyword.kwlist)
```

# Problem-22
### Write a Python program to print numbers from 1 to N.
### Take an integer N as input from the user.
### Initialize a variable starting from 1.
### Use a while loop to iterate until the value reaches N.
### Print each number during every loop iteration.


```python
n = int(input())
i = 1

while i <= n:
    print(i)
    i += 1
```
