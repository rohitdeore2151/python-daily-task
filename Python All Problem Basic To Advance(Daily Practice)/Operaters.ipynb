### Write a Python program to accept principal, rate, and time.
### Calculate compound interest using CI = P(1 + R/100)^T − P.
### Display the compound interest.
### Use proper mathematical operations.

```python
p = int(input())
r = int(input())
t = int(input())

ci = p * (1 + r / 100) ** t - p

print(int(ci))
```
