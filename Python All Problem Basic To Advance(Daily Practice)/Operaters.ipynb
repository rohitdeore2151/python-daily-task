# Problem-12
### Write a Python program to accept values of a, b, and c.
### Calculate discriminant using D = b² − 4ac/2a
### Find and display the roots.
### Handle all possible cases.

```python
import cmath

a = float(input())
b = float(input())
c = float(input())

d = b ** 2 - 4 * a * c

root1 = (-b + cmath.sqrt(d)) / (2 * a)
root2 = (-b - cmath.sqrt(d)) / (2 * a)

print(root1, root2)
```

# Problem-15
### Write a Python program to accept three numbers.
### Calculate the mean using (n1 + n2 + n3) / 3.
### Find the median after sorting the numbers.
### Determine the mode or state no mode exists.

```python
n1 = int(input())
n2 = int(input())
n3 = int(input())

numbers = [n1, n2, n3]
numbers.sort()

mean = (n1 + n2 + n3) / 3
median = numbers[1]

if n1 == n2 or n1 == n3:
    mode = n1
elif n2 == n3:
    mode = n2
else:
    mode = None

print(f"Mean={mean:.2f}")
print(f"Median={median}")

if mode is not None:
    print(f"Mode={mode}")
else:
    print("Mode=No Mode")
```
