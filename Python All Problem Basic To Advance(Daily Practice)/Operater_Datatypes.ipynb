# Problem-1
### Write a Python program to read a sentence and a word.
### Check whether the word exists in the sentence.
### Use the in membership operator.
### Print True or False accordingly.

```python
sentence = input()
word = input()

print(word in sentence)
```

# Problem-2
### Write a Python program to read a list of numbers and a number.
### Check whether the number is not present in the list using not in.
### Print True if it is not found.
### Otherwise, print False.

```python
numbers = list(map(int, input().split()))
num = int(input())

print(num not in numbers)
```

# Problem-3
### Write a Python program to read a tuple of characters and one character.
### Check whether the character exists in the tuple using in.
### Print True if found.
### Otherwise, print False.


```python
characters = tuple(input().split())
ch = input()

print(ch in characters)
```

# Problem-4
### Write a Python program to read a string.
### Check whether at least one vowel exists in the string.
### Consider both uppercase and lowercase vowels.
### Print True or False.

```python

s = input()

vowels = "aeiouAEIOU"

result = any(ch in vowels for ch in s)

print(result)
```

# Problem-5
### Write a Python program to read an email address.
### Check whether @gmail.com exists in the email string.
### Print Valid if it exists.
### Otherwise, print Invalid

```python

email = input()

if "@gmail.com" in email:
    print("Valid")
else:
    print("Invalid")
```
