# Problem-12
### Write a Python program to read a password.
### Check if it contains at least one digit.
### Also check for at least one special character.
### Print Strong or Weak.

```python

password = input()

digits = "0123456789"
special = "!@#$%^&*"

has_digit = any(ch in digits for ch in password)
has_special = any(ch in special for ch in password)

if has_digit and has_special:
    print("Strong")
else:
    print("Weak")
```

# Problem-18
### Write a Python program to read an integer.
### Add 1 to the number without using the + operator.
### Use only bitwise operations to perform addition.
### Ensure the value increases correctly by one.
### Print the result.

```python

n = int(input())

mask = 1

while n & mask:
    n = n ^ mask
    mask = mask << 1

n = n ^ mask

print(n)
```
