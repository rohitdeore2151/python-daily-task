# Problem-1
### Write a Python program to read a sentence and a word.
### Check whether the word exists in the sentence.
### Use the in membership operator.
### Print True or False accordingly.

```python
sentence = input()
word = input()

print(word in sentence)
```

# Problem-2
### Write a Python program to read a list of numbers and a number.
### Check whether the number is not present in the list using not in.
### Print True if it is not found.
### Otherwise, print False.

```python
numbers = list(map(int, input().split()))
num = int(input())

print(num not in numbers)
```

# Problem-3
### Write a Python program to read a tuple of characters and one character.
### Check whether the character exists in the tuple using in.
### Print True if found.
### Otherwise, print False.


```python
characters = tuple(input().split())
ch = input()

print(ch in characters)
```

# Problem-4
### Write a Python program to read a string.
### Check whether at least one vowel exists in the string.
### Consider both uppercase and lowercase vowels.
### Print True or False.

```python

s = input()

vowels = "aeiouAEIOU"

result = any(ch in vowels for ch in s)

print(result)
```

# Problem-5
### Write a Python program to read an email address.
### Check whether @gmail.com exists in the email string.
### Print Valid if it exists.
### Otherwise, print Invalid

```python

email = input()

if "@gmail.com" in email:
    print("Valid")
else:
    print("Invalid")
```

# Problem-6
### Write a Python program to read a password string.
### Check if it contains any special character from !@#$%^&*.
### If found, print True.
### Otherwise, print False.

```python

password = input()

special = "!@#$%^&*"

result = any(ch in special for ch in password)

print(result)
```

# Problem-7
### Write a Python program to read a list of string values.
### Check whether the value "None" exists in the list.
### Use the in operator for checking.
### Print True or False

```python

# Read a list of string values
values = input("Enter string values separated by spaces: ").split()

# Check whether "None" exists in the list
result = "None" in values

# Print True or False
print(result)
```

# Problem-8
### Write a Python program to read a string.
### Check whether all five vowels (a, e, i, o, u) are present.
### Use membership operators for checking.
### Print True or False.

```python

s = input()

vowels = "aeiou"

result = all(ch in s.lower() for ch in vowels)

print(result)
```

# Problem-9
### Write a Python program to read an email address.
### Check whether it contains @ and ends with .com.
### If both conditions are true, print Valid.
### Otherwise, print Invalid.

```python

email = input()

if "@" in email and email.endswith(".com"):
    print("Valid")
else:
    print("Invalid")
```
