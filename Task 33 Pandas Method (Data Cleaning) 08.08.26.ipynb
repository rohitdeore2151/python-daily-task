#### Q9. Fix Age

#### Q10. Fix Annual Income
#### Clean Annual_Income.
#### Requirements:
#### 1. Convert invalid values to missing.
#### 2. Convert the column to a numeric datatype.
#### 3. Calculate the median.
#### 4. Fill missing values using the median.
### Use: pd.to_numeric(), median(), fillna(), astype()

```python
df['Annual_Income'] = pd.to_numeric(df['Annual_Income'],errors = 'coerce')

df["Annual_income"] = df ["Annual_Income"].astype("float")

median_income = df['Annual_Income'].median()

df['Annual_Income'] = df['Annual_Income'].fillna(median_income)

print(df[["Annual_Income"]].head())
print(df["Annual_Income"].dtype)
print(df["Annual_Income"].isna().sum())
```

## Part 3 — Numerical Cleaning
#### Q11. Clean Account Balance
#### The Account_Balance column contains invalid strings and missing values.
#### Clean the column and replace missing values with the mean Account Balance.
#### Use:
#### • pd.to_numeric()
#### • mean()
#### • fillna()

```python
df["Account_Balance"] = pd.to_numeric(df["Account_Balance"],errors = "coerce")

df["Account_Balance"] = df["Account_Balance"].astype(float)

mean_balance = df["Account_Balance"].mean()

df["Account_Balance"] = df["Account_Balance"].fillna(mean_balance)

print(df["Account_Balance"].head)
print(df["Account_Balance"].dtype)
print(df[["Account_Balance"]].isna().sum())
```

#### Q21. Clean Active
#### Active is supposed to be a binary column, but it contains:
#### Yes
#### No
#### Y
#### N
#### YES
#### NO
#### yes
#### no
#### 1
#### 0
#### Transform these into a consistent representation:
#### Yes
#### No
#### Then verify the result with value_counts().
#### Use: replace(), value_counts()

```python
df["Active"] = df["Active"].astype(str).str.strip().str.upper()

df["Active"] = df["Active"].replace({
    "Y": "Yes",
    "YES": "Yes",
    "1": "Yes",

    "N": "No",
    "NO": "No",
    "0": "No"
})

print(df["Active"].value_counts(dropna=False))
```

## Part 7 — Create New Columns
#### Q26. Create Annual Income in Lakhs
#### Create:
#### Income_Lakhs
#### using:
#### AnnualIncome / 100000
#### Then check the first 10 records.

```python
df["Annual_Income"] = pd.to_numeric(df["Annual_Income"], errors="coerce")
df["Annual_Income"] = df["Annual_Income"].fillna(df["Annual_Income"].median())

df["Income_Lakhs"] = df["Annual_Income"] / 100000
print(df[["Annual_Income", "Income_Lakhs"]].head(10))
```

#### Q27. Create Monthly Income
#### Create:
#### MonthlyIncome
#### using:
#### AnnualIncome / 12
#### Then create:
#### Income_After_Spending
#### using:
#### MonthlyIncome - MonthlySpend

```python
df["Annual_Income"] = pd.to_numeric(df["Annual_Income"], errors="coerce")
df["Monthly_Spend"] = pd.to_numeric(df["Monthly_Spend"], errors="coerce")

df["MonthlyIncome"] = df["Annual_Income"] / 12

df["Income_After_Spending"] = (df["MonthlyIncome"] - df["Monthly_Spend"])

print(df[["Annual_Income","MonthlyIncome","Monthly_Spend","Income_After_Spending"]].head(10))
```

#### Q28. Create Balance-to-Income Ratio
#### Create:
#### Balance_Income_Ratio
#### using:
#### AccountBalance / AnnualIncome
#### Then inspect the first 10 rows.
#### Think carefully about division by zero and missing values.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("C:\\Users\\rohit\\Downloads\\customer_analytics.csv")

df["Annual_Income"] = pd.to_numeric(df["Annual_Income"], errors="coerce")
df["Account_Balance"] = pd.to_numeric(df["Account_Balance"], errors="coerce")

df["Annual_Income"] = df["Annual_Income"].replace(0, np.nan)

df["Balance_Income_Ratio"] = (df["Account_Balance"] / df["Annual_Income"])

print(df[["Account_Balance","Annual_Income","Balance_Income_Ratio"]].head(10))
```
