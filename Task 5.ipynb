```python
# 21. (Hard) Create a dictionary student with 4 key-value pairs. 
# Then use update() to add another dictionary, remove the last item using popitem(), 
# and finally print all items using items().

student = {
    "Name": "Rohit",
    "Age": 22,
    "Course": "B-Tech",
    "Marks": 80
}

Add = {
    "City": "Pune",
    "Grade": "A"
}

student.update(Add)

student.popitem()

print("Student Dictionary:")

for key, value in student.items():
    print(key, ":", value)
```

```python
# 22. (Hard) Create a tuple with 8 numbers. Find the index of the number 25.
# If it exists, count how many times it appears.
# Then convert the tuple into a list, sort it in descending order, and print.

numbers = (10, 25, 40, 25, 60, 15, 80, 25)

if 25 in numbers:
    print("Index of 25:", numbers.index(25))
    print("Count of 25:", numbers.count(25))
else:
    print("25 not found in the tuple.")

num_list = list(numbers)

num_list.sort(reverse=True)

print("List in descending order:", num_list)
```

```python
# 23. (Hard) Take 4 student names and their marks as input from the user.
# Store them in a dictionary (name as key, marks as value). Then find and
# print the student with the highest marks.

students = {}

for i in range(4):
    name = input("Enter student name: ")
    marks = int(input("Enter marks: "))
    students[name] = marks

highest_student = max(students, key=students.get)

print("Student with the highest marks:")
print(highest_student, ":", students[highest_student])
```

```python
# 24. (Hard) Create a dictionary info = {"name": "Aman", "age": 21, "city":"Mumbai"}. 
# Make a copy of it. Clear the original dictionary. 
# Then add a new key "country": "India" in the copy and print both dictionaries.

info = {
    "name": "Aman",
    "age": 21,
    "city": "Mumbai"
}

info_copy = info.copy()

info.clear()

info_copy["country"] = "India"

print("Original Dictionary:", info)
print("Copied Dictionary:", info_copy)
```

```python
# 25. (Hard) Given data = (10, 20, 30, 20, 40, 20, 50), convert it into 
# adictionary where key is the number and value is its count of occurrences
# (using tuple methods + dictionary)

data = (10, 20, 30, 20, 40, 20, 50)

count_dict = {}

for num in data:
    count_dict[num] = data.count(num)

print(count_dict)
```
