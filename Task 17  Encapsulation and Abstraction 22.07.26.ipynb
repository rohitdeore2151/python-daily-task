#### 1. Class Name: BankAccount
#### Problem
#### Create a class named BankAccount.
#### Explanation
#### A bank account stores money. The balance should not be directly accessible from outside
#### the class, so create a private attribute named __balance.
#### Implement the following methods:
#### • deposit(amount)
#### o Deposit money into the account.
#### o Amount must be greater than 0.
#### • withdraw(amount)
#### o Withdraw money.
#### o Withdrawal should happen only if:
#### ▪ amount > 0
#### ▪ amount <= available balance

```python
class BankAccount:

    def __init__(self):
        self.__balance = 0

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"₹{amount} deposited successfully.")
        else:
            print("Invalid deposit amount.")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount.")
        elif amount > self.__balance:
            print("Insufficient balance.")
        else:
            self.__balance -= amount
            print(f"₹{amount} withdrawn successfully.")

    def get_balance(self):
        return self.__balance


# Driver Code
account = BankAccount()

account.deposit(5000)
account.deposit(-100)

account.withdraw(1000)
account.withdraw(6000)

print("Current Balance:", account.get_balance())
```

#### Create a class named Employee.

#### Salary should not be modified directly.

#### • get_salary()

#### Current Salary = 50,000

```python
class Employee:

    def __init__(self, salary):
        self.__salary = salary

    def set_salary(self, new_salary):
        minimum_salary = self.__salary * 0.90

        if new_salary >= minimum_salary:
            self.__salary = new_salary
            print("Salary updated successfully.")
        else:
            print("Salary cannot decrease by more than 10%.")

    def get_salary(self):
        return self.__salary


# Driver Code
employee = Employee(50000)

print("Current Salary:", employee.get_salary())

employee.set_salary(48000)
print("Updated Salary:", employee.get_salary())

employee.set_salary(45000)
print("Updated Salary:", employee.get_salary())

employee.set_salary(40000)
print("Final Salary:", employee.get_salary())
```

#### • get_model()

```python
from datetime import datetime

class Car:

    def __init__(self):
        self.__model = ""
        self.__year = 0

    def set_model(self, model):
        self.__model = model
        print("Model updated successfully.")

    def get_model(self):
        return self.__model

    def set_year(self, year):
        current_year = datetime.now().year

        if year <= current_year:
            self.__year = year
            print("Year updated successfully.")
        else:
            print("Invalid year! Year cannot be greater than the current year.")

    def get_year(self):
        return self.__year


# Driver Code
car = Car()

car.set_model("Toyota Fortuner")
car.set_year(2024)

print("Model:", car.get_model())
print("Year:", car.get_year())

car.set_year(2030)

print("Final Year:", car.get_year())
```

#### 10. Class Name: Account
#### Problem
#### Create class Account.
#### Explanation
#### Private attribute
#### __balance
#### Whenever balance changes,
#### store every balance value inside a class variable
#### Example
#### history = []
#### Methods
#### • deposit()
#### • withdraw()
#### Every successful update should append the latest balance to history.

```python
class Account:

    # Class variable
    history = []

    def __init__(self):
        self.__balance = 0

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            Account.history.append(self.__balance)
            print("Amount deposited successfully.")
        else:
            print("Invalid deposit amount.")

    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            Account.history.append(self.__balance)
            print("Amount withdrawn successfully.")
        else:
            print("Invalid withdrawal.")

    def get_balance(self):
        return self.__balance


# Driver Code
acc = Account()

acc.deposit(1000)
acc.deposit(500)
acc.withdraw(300)
acc.withdraw(2000)

print("Current Balance:", acc.get_balance())
print("History:", Account.history)
```

#### 16. Class Name: Shape
#### Problem
#### Create an abstract class Shape.
#### Explanation
#### Declare abstract methods
#### • calculate_area()
#### • calculate_perimeter()
#### Create subclasses like
#### • Circle
#### • Rectangle
#### Implement both methods.

```python
from abc import ABC, abstractmethod

# Abstract Class
class Shape(ABC):

    @abstractmethod
    def calculate_area(self):
        pass

    @abstractmethod
    def calculate_perimeter(self):
        pass


# Circle Class
class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return 3.14 * self.radius * self.radius

    def calculate_perimeter(self):
        return 2 * 3.14 * self.radius


# Rectangle Class
class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        return self.length * self.width

    def calculate_perimeter(self):
        return 2 * (self.length + self.width)


# Driver Code
circle = Circle(5)

print("Circle Area:", circle.calculate_area())
print("Circle Perimeter:", circle.calculate_perimeter())

print()

rectangle = Rectangle(10, 5)

print("Rectangle Area:", rectangle.calculate_area())
print("Rectangle Perimeter:", rectangle.calculate_perimeter())
```

#### 17. Class Name: Vehicle
#### Problem
#### Create abstract class Vehicle.
#### Explanation
#### Abstract methods
#### • start()
#### • stop()
#### Implement in
#### • Car
#### • Bike
#### Each subclass should define how the vehicle starts and stops.

```python
from abc import ABC, abstractmethod

# Abstract Class
class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass


# Car Class
class Car(Vehicle):

    def start(self):
        print("Car starts with a key or push button.")

    def stop(self):
        print("Car stops by applying brakes.")


# Bike Class
class Bike(Vehicle):

    def start(self):
        print("Bike starts with a self-start or kick.")

    def stop(self):
        print("Bike stops by applying brakes.")


# Driver Code
car = Car()
bike = Bike()

print("Car:")
car.start()
car.stop()

print("\nBike:")
bike.start()
bike.stop()
```

#### 18. Class Name: Payment
#### Problem
#### Create abstract class Payment.
#### Explanation
#### Abstract method
#### process_payment(amount)
#### Create subclasses
#### • CreditCardPayment
#### • UPIPayment
#### Each should process payment differently.

```python
from abc import ABC, abstractmethod

# Abstract Class
class Payment(ABC):

    @abstractmethod
    def process_payment(self, amount):
        pass


# Credit Card Payment
class CreditCardPayment(Payment):

    def process_payment(self, amount):
        print(f"₹{amount} paid using Credit Card.")


# UPI Payment
class UPIPayment(Payment):

    def process_payment(self, amount):
        print(f"₹{amount} paid using UPI.")


# Driver Code
credit = CreditCardPayment()
upi = UPIPayment()

credit.process_payment(2500)
upi.process_payment(1500)
```

#### 19. Class Name: Animal
#### Problem
#### Create abstract class Animal.
#### Explanation
#### Abstract method
#### make_sound()
#### Create subclasses
#### • Dog
#### • Cat
#### Each returns its own sound.

```python
from abc import ABC, abstractmethod

# Abstract Class
class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        pass


# Dog Class
class Dog(Animal):

    def make_sound(self):
        return "Bark"


# Cat Class
class Cat(Animal):

    def make_sound(self):
        return "Meow"


# Driver Code
dog = Dog()
cat = Cat()

print("Dog:", dog.make_sound())
print("Cat:", cat.make_sound())
```

#### 20. Class Name: Report
#### Problem
#### Create abstract class Report.
#### Explanation
#### Abstract method
#### generate_report()
#### Create
#### • PDFReport
#### • ExcelReport
#### Each generates report differently.

```python
from abc import ABC, abstractmethod

# Abstract Class
class Report(ABC):

    @abstractmethod
    def generate_report(self):
        pass


# PDF Report
class PDFReport(Report):

    def generate_report(self):
        print("Generating PDF Report...")


# Excel Report
class ExcelReport(Report):

    def generate_report(self):
        print("Generating Excel Report...")


# Driver Code
pdf = PDFReport()
excel = ExcelReport()

pdf.generate_report()
excel.generate_report()
```
