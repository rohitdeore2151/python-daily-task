```python
# Parent 1
class Person:

    # Constructor
    def __init__(self, name, age, **kwargs):
        super().__init__(**kwargs)
        self.__Name = name
        self.__Age = age

        # Getter
        def get_name(self):
            return f"Person Name : {self.__Name}"
    
        def get_age(self):
            return f"Person Age : {self.__Age}"

    # Static Method
    @staticmethod
    def valid_age(age):
        return 5 <= age <= 100

    # Setter
    def set_age(self, age):
        if Person.valid_age(age):
            self.__Age = age
            return f"Age Updated : {self.__Age}"
        else:
            return "Invalid Age"

    # Show
    def show(self):
        print(self.get_name())
        print(self.get_age())


# Parent 2
class Academic:

    # Constructor
    def __init__(self, subject, grade, **kwargs):
        super().__init__(**kwargs)
        self.__Subject = subject
        self.__Grade = grade

    # Getter
    def get_subject(self):
        return f"Subject Name : {self.__Subject}"

    def get_grade(self):
        return f"Grade : {self.__Grade}"

    # Static Method
    @staticmethod
    def get_letter(grade):
        if 90 <= grade <= 100:
            return "A+"
        elif 80 <= grade <= 89:
            return "A"
        elif 70 <= grade <= 79:
            return "B"
        elif 60 <= grade <= 69:
            return "C"
        elif 50 <= grade <= 59:
            return "D"
        else:
            return "Fail"

    # Setter
    def set_grade(self, grade):
        if 0 <= grade <= 100:
            self.__Grade = grade
            return f"Grade Updated : {self.__Grade}"
        else:
            return "Please enter grade between 0 and 100"


# Child Class
class Student(Person, Academic):

    school_name = "NareshIT"
    total_students = 0

    # Constructor
    def __init__(self, name, age, subject, grade, std_id, fees):

        super().__init__(
            name=name,
            age=age,
            subject=subject,
            grade=grade
        )

        self.__Student_ID = std_id
        self.__Fees = fees

        Student.total_students += 1

    # Getter
    def get_std_id(self):
        return f"Student ID : {self.__Student_ID}"

    def get_fees(self):
        return f"Student Fees : {self.__Fees}"

    # Static Method
    @staticmethod
    def valid_fees(fees):
        return fees > 0

    # Setter
    def set_fees(self, fees):
        if Student.valid_fees(fees):
            self.__Fees = fees
            return f"Fees Updated : {self.__Fees}"
        else:
            return "Invalid Fees"

    # Class Method
    @classmethod
    def get_total_students(cls):
        return cls.total_students

    @classmethod
    def set_school_name(cls, name):
        cls.school_name = name
        return f"School Name Updated : {cls.school_name}"

    # Show
    def show(self):
        print(self.get_std_id())
        print(self.get_fees())
        print(self.get_name())
        print(self.get_age())
        print(self.get_subject())
        print(self.get_grade())
        print(f"Letter Grade : {Academic.get_letter(int(self.get_grade().split(': ')[1]))}")


# Object
std1 = Student("Rohit",22,"Math", 95,"STD123",5000)

std1.show()

```

```python
# Parent Class
class Employee:

    # Class Attribute
    total_emp = 0

    # Constructor
    def __init__(self, name, salary, dept):
        self.__name = name
        self.__salary = salary
        self.__dept = dept
        Employee.total_emp += 1

    # Getters
    def get_name(self):
        return self.__name

    def get_salary(self):
        return self.__salary

    def get_dept(self):
        return self.__dept

    # Static Method
    @staticmethod
    def valid_salary(salary):
        return salary > 0

    # Setters
    def set_salary(self, salary):
        if Employee.valid_salary(salary):
            self.__salary = salary
            return "Salary Updated"
        else:
            return "Invalid Salary"

    # Class Method
    @classmethod
    def get_total(cls):
        return cls.total_emp

    # Method
    def work(self):
        print("Employee is working.")

    # Bonus (5%)
    def calculate_bonus(self):
        return self.__salary * 0.05

    # Show
    def show(self):
        print("Name      :", self.get_name())
        print("Department:", self.get_dept())
        print("Salary    :", self.get_salary())
        print("Bonus     :", self.calculate_bonus())


# Child 1
class Developer(Employee):

    def __init__(self, name, salary, dept, language):
        super().__init__(name, salary, dept)
        self.__language = language

    def get_language(self):
        return self.__language

    # Override
    def work(self):
        print("Developer is writing code.")

    # Bonus 20%
    def calculate_bonus(self):
        return self.get_salary() * 0.20

    def show(self):
        super().show()
        print("Language  :", self.get_language())


# Child 2
class Designer(Employee):

    def __init__(self, name, salary, dept, tool):
        super().__init__(name, salary, dept)
        self.__tool = tool

    def get_tool(self):
        return self.__tool

    # Override
    def work(self):
        print("Designer is creating UI.")

    # Bonus 15%
    def calculate_bonus(self):
        return self.get_salary() * 0.15

    def show(self):
        super().show()
        print("Tool      :", self.get_tool())


# Child 3
class Manager(Employee):

    def __init__(self, name, salary, dept, team_size):
        super().__init__(name, salary, dept)
        self.__team_size = team_size

    def get_team_size(self):
        return self.__team_size

    # Override
    def work(self):
        print("Manager is managing team.")

    # Bonus 20%
    def calculate_bonus(self):
        return self.get_salary() * 0.20

    def show(self):
        super().show()
        print("Team Size :", self.get_team_size())


# Objects
dev = Developer("Rohit", 50000, "IT", "Python")
des = Designer("Priya", 45000, "Design", "Figma")
mgr = Manager("Amit", 70000, "HR", 12)

print("------ Developer ------")
dev.show()
dev.work()

print("\n------ Designer ------")
des.show()
des.work()

print("\n------ Manager ------")
mgr.show()
mgr.work()

print("\nTotal Employees:", Employee.get_total())
```
