#### 13-18: Positional Arguments & Keyword Arguments

#### 13. Write a function full_name(first, last) and call it using positional arguments.

```python
def full_name(first,last):
    print("Full Name:", first,last)

full_name("Rohit","Deore")
full_name("Deore","Rohit")
```

#### 14. Call the same function using keyword arguments.

```python
def full_name(first,last):
    print("Full Name:", first,last)

full_name(first = "Rohit" , last = "Deore")
full_name(last = "Deore" , first = "Rohit")
```

#### 16. Create a function calculate(a, b, operation) and show how keyword args improve readability.

```python
def calculate(a, b, operation):
    if operation == "add":
        print("Result:", a + b)
    elif operation == "subtract":
        print("Result:", a - b)
    elif operation == "multiply":
        print("Result:", a * b)
    elif operation == "divide":
        if b != 0:
            print("Result:", a / b)
        else:
            print("Cannot divide by zero.")
    else:
        print("Invalid operation.")

calculate(10, 5, "add")

calculate(a=10, b=5, operation="multiply")
calculate(operation="subtract", a=20, b=8)
```

#### 17. Write a function order(item, quantity, price) and call it with mixed positional + keyword.

### 19-25: args (Multiple Positional Arguments)

#### 19. Write a function add_all(*args) that returns sum of all numbers passed.

```python
def add_all(*args):
    return sum(args)

print(add_all(10, 20, 30, 40))
```

```python
#20. Create a function print_all(*args) that prints all arguments.

def print_all(*args):
    for item in args:
        print(item)

# Function call
print_all(10, 20, 30)
print_all("Python", "Java", "C++")
```
