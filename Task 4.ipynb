## Practical Questions on String Methods & Lists with Difficulty



```python
# 1.Take a string from user and replace all spaces with underscore _ using replace() method.
```

```python
string = input("Enter any string")
string = string.replace(" ","_")
print(string)

```

```python
# 2. (Easy) Take any string and remove leading and trailing spaces using strip().
```

```python
string = input("Enter any string")
string = string.strip()
print(string)
```

```python
# '3. (Easy) Take a sentence and check whether it starts with "Hello" using startswith().
```

```python
string = input("Enter the string")
string = string.startswith("Hello")
print(string)
```

```python
# 4. (Easy) Take a string and check if it ends with ".com" using endswith().
```

```python
string = input("Enter the string")
string = string.endswith(".com")
print(string)
```

```python
# 5. (Easy) Take a comma-separated string like "apple,banana,mango" and split it using split().
```

```python
string = input("apple,banana,mango")
string = string.split(",")
print(string)
```

```python
# 6. (Easy) Create a list of 5 fruits and join them using space " " with join().
```

```python
text = ["banana","apple","mango","guava","stobery"]
text = ' '.join(text)
print(text)

```

```python
# 7. (Easy) Create a list: [" Python ", " is ", " awesome "]. Strip each word and join them.
```

```python
list = [" Python ", " is ", " awesome "]

list[0]=list[0].strip()
list[1]=list[1].strip()
list[2]=list[2].strip()

list=("-".join(list))

print(list)
```

```python
#12. (Medium) Create a list of 5 city names from user input. Join them with new line "\n" and print.
```

```python
cities = []
for i in range(5):
    city = input(f"Enter city {i+1}: ")
    cities.append(city)
    
result = "\n".join(cities)

print("City Names:")
print(result)
```
