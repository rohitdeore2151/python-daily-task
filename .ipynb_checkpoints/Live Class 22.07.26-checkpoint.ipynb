```python
# with capsulation:

class Bank_account:
    #constructor
    def __init__(self,owner,bal):

        #create the attributes
        self.owner_name =str(owner)
        self.__Bank_balance = int(bal) #private 


    #getter : Access the private attribute:

    def get_balance(self):
        return self.__Bank_balance

    #setter: set_validation

    def set_balance(self,amt):
        if amt > 0:
            self.__Bank_balance += amt

            return self.__Bank_balance

        else:
            return " Please Enter The Valid Ammount"
        #method 1:
    def display_bank_details(self):
        print(f"owner Name :{self.owner_name}")
        print(f"bank Balance :{self.Bank_balance}")
# create the class object:

bank1 = Bank_account("Rohit",2000)

#call the method

bank1.set_balance(222)
```

```python
from abc import ABC, abstractmethod

# food management:

class Food(ABC):

    # constructor :
    def __init__(self,f_name,f_price):
        self.__Food_name = f_name
        self.__Food_price = f_price

    # getter name
    def get_food_name(self):
        return self.__food_name

    # getter price
    def get_food_price(self):
        return self.__Food_price

    # abstract method

    @abstractmethod
    def prepare(self):
        pass
    # abstract Method:

    @abstractmethod
    def deliver(self):
        pass

        # display all details
        def display_all_details(self):
            print(f"Food Name : {self.__Food_name}")
            print(f"Food Price : {self.__Food_price}")
            self.prepare()
            self.delever()


# child class

class Pizza(Food):

    def prepare(self):
        print("pizza is prepared")

    def deliver(self):
        print("pizza is Delivered")

class Burger(Food):

    def prepare(self):
        print("Burger is prepared")

    def deliver(self):
        print("Burger is Delivered")

class Biryani(Food):

    def prepare(self):
        print("Biryani is prepared")

    def deliver(self):
        print("Biryani is Delivered")

# creat the class object

pizza = Pizza("Pizza",500)
burger = Burger("Burger",200)
biryani = Biryani("Biryani",700)

Pizza.display_all_details()
```

```python
from abc import ABC, abstractmethod

# Parent Abstract Class
class Parent(ABC):
    def __init__(self, name, designation):
        self.__name = name               # Private data
        self.__designation = designation # Private data

    # Getter methods
    def get_name(self):
        return self.__name

    def get_designation(self):
        return self.__designation

    # Abstract methods
    @abstractmethod
    def work(self):
        pass

    @abstractmethod
    def salary(self):
        pass

    # Display method
    def display(self):
        print("Name :", self.get_name())
        print("Role :", self.get_designation())
        print("Work :", self.work())
        print("Salary :", self.salary())


# Child Class 1
class Employee(Parent):
    def work(self):
        return "Developing Machine Learning Models"

    def salary(self):
        return "₹60,000 per month"


# Child Class 2
class Manager(Parent):
    def work(self):
        return "Managing Data Science Team"

    def salary(self):
        return "₹1,20,000 per month"


# Driver Code
emp = Employee("Rohit", "Data Scientist")
mgr = Manager("Priya", "Data Science Manager")

print("Employee Details")
emp.display()

print("\nManager Details")
mgr.display()
```
