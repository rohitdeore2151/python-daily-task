# Shallow Copy vs Deep Copy and Class, Objects

#### 1. Create a list original = [1, 2, [3, 4], 5]. Make a shallow copy named shallow and change shallow[2][0] = 99.
#### Print both lists and explain the result.

```python
import copy

original = [1,2,[3,4],5]

shallow = copy.copy(original)

shallow[2][0] = 90

print("original list",original)
print("shallow",shallow)
```

#### 2. Do the same as above but use deepcopy instead of shallow copy. Show the difference.

```python
import copy

original = [1,2,[3,4],5]

shallow = copy.deepcopy(original)

shallow[2][0] = 90

print("original list",original)
print("shallow",shallow)
```

#### 3. Write a program to demonstrate shallow copy using copy.copy() on a nested list.

```python
import copy

original = [[1, 2], [3, 4], [5, 6]]

shallow_copy = copy.copy(original)

print("Before modification:")
print("Original:", original)
print("Shallow Copy:", shallow_copy)

shallow_copy[0][0] = 100

print("\nAfter modifying an inner list in shallow copy:")
print("Original:", original)
print("Shallow Copy:", shallow_copy)

shallow_copy[1] = ["A", "B"]

print("\nAfter replacing an outer list element in shallow copy:")
print("Original:", original)
print("Shallow Copy:", shallow_copy)
```

#### 4. Write a program to demonstrate deep copy using copy.deepcopy() on the same nested list.

```python
import copy

original = [[1, 2], [3, 4], [5, 6]]

shallow_copy = copy.deepcopy(original)

print("Before modification:")
print("Original:", original)
print("Shallow Copy:", shallow_copy)

shallow_copy[0][0] = 100

print("\nAfter modifying an inner list in shallow copy:")
print("Original:", original)
print("Shallow Copy:", shallow_copy)

shallow_copy[1] = ["A", "B"]

print("\nAfter replacing an outer list element in shallow copy:")
print("Original:", original)
print("Shallow Copy:", shallow_copy)
```

#### 6. Create a class Person with a list attribute hobbies. Create two objects using shallow copy and show how changing hobbies affects both.

```python
import copy

class Person:
    def __init__(self, hobbies):
        self.hobbies = hobbies

person1 = Person(["Reading", "writing"])

person2 = copy.copy(person1)

print("Before modification:")
print("Person1 hobbies:", person1.hobbies)
print("Person2 hobbies:", person2.hobbies)

person2.hobbies.append("Gaming")

print("\nAfter modification:")
print("Person1 hobbies:", person1.hobbies)
print("Person2 hobbies:", person2.hobbies)
```

#### 7. Write a function manual_deep_copy(lst) that manually creates a deep copy of a nested list without using copy.deepcopy().

```python
def manual_deep_copy(lst):
    new_list = []

    for item in lst :
        if isinstance(item,list):
            new_list.append(manual_deep_copy(item))

        else:
            new_list.append(item)

    return new_list

original = [1,[2,3],[4,[5,6]]]

copied = manual_deep_copy(original)
print(copied)

copied[1].append(100)
copied[2][1].append(200)

print("Original:",original)
print("Copied:",copied)
    
```

#### 9. (Hard) Create a function that detects whether a copy is shallow or deep by modifying a nested element and checking the original.

```python
import copy

def detect_copy(original, copied):
    old_value = original[1][0]

    copied[1][0] = 999

    if original[1][0] == 999:
        result = "Shallow Copy"
    else:
        result = "Deep Copy"

    original[1][0] = old_value

    return result

data = [10, [20, 30], {"name": "Ram"}, 40]

shallow = copy.copy(data)
deep = copy.deepcopy(data)

print("Shallow copy detected as:", detect_copy(data, shallow))
print("Deep copy detected as:", detect_copy(data, deep))
```

#### 10. (Hard) Create a class Student with attributes name and marks (list). Demonstrate what
#### happens when you do shallow copy vs deep copy of a Student object.

```python
import copy

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(f"Name: {self.name}, Marks: {self.marks}")

student1 = Student("Rohit", [85, 90, 95])

student2 = copy.copy(student1)      
student3 = copy.deepcopy(student1)  


student1.name = "Rohan"
student1.marks.append(100)

print("Original Student:")
student1.display()

print("\nShallow Copy:")
student2.display()

print("\nDeep Copy:")
student3.display()
```

#### 5. Bank Account
#### Attributes
#### • account_holder
#### • balance
#### Methods
#### • deposit()
#### • withdraw()
#### Withdraw should happen only if balance is sufficient.

```python
class BankAccount:
    
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance!")

    def display(self):
        print("Account Holder:", self.account_holder)
        print("Balance:", self.balance)


acc1 = BankAccount("John", 5000)

print("Initial Account Details:")
acc1.display()

acc1.deposit(2000)

acc1.withdraw(3000)

acc1.withdraw(5000)

print("\nFinal Account Details:")
acc1.display()
```

#### 6. Car Speed

#### Method
#### increase_speed(value)
#### Maximum speed should never exceed 200 km/h.

```python
class Car:

    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
    
    def increase_speed(self, value):
        self.speed += value
        if self.speed > 200:
            self.speed = 200

    def display(self):
        print("Brand:", self.brand)
        print("Speed:", self.speed, "km/h")


car1 = Car("Toyota", 180)

print("Before Increasing Speed:")
car1.display()

car1.increase_speed(30)

print("\nAfter Increasing Speed:")
car1.display()
```

#### 7. Temperature
#### Store temperature.
#### Method
#### set_temperature()
#### Condition
#### Temperature should always remain between
#### -20°C
#### to
#### 50°C
#### Otherwise print Invalid.

```python
class Temperature:
    
    def __init__(self, temperature):
        self.temperature = temperature

    def set_temperature(self, value):
        if -20 <= value <= 50:
            self.temperature = value
            print("Temperature set to", self.temperature, "°C")
        else:
            print("Invalid")

    def display(self):
        print("Current Temperature:", self.temperature, "°C")


t1 = Temperature(25)

print("Initial Temperature:")
t1.display()

t1.set_temperature(40)

t1.set_temperature(60)

print("\nFinal Temperature:")
t1.display()
```

#### 8. Movie Ticket

#### Attributes
#### • movie_name
#### • seats
#### Method
#### book_ticket(number)
#### Booking should happen only if seats are available.

```python
class MovieTicket:
    # Constructor
    def __init__(self, movie_name, seats):
        self.movie_name = movie_name
        self.seats = seats

    def book_ticket(self, number):
        if number <= self.seats:
            self.seats -= number
            print(number, "ticket(s) booked successfully.")
        else:
            print("Seats not available!")

    def display(self):
        print("Movie Name:", self.movie_name)
        print("Available Seats:", self.seats)


m1 = MovieTicket("Pushpa 2", 50)

print("Before Booking:")
m1.display()

m1.book_ticket(20)

m1.book_ticket(40)

print("\nAfter Booking:")
m1.display()
```

#### 9. Library
#### Attributes
#### • book_name
#### • available_copies
#### Method
#### borrow_book()
#### Reduce copy only if available.

```python
class Library:

    def __init__(self, book_name, available_copies):
        self.book_name = book_name
        self.available_copies = available_copies

    def borrow_book(self):
        if self.available_copies > 0:
            self.available_copies -= 1
            print("Book borrowed successfully.")
        else:
            print("No copies available!")

    def display(self):
        print("Book Name:", self.book_name)
        print("Available Copies:", self.available_copies)


book1 = Library("Python Programming", 3)

print("Before Borrowing:")
book1.display()

book1.borrow_book()
book1.borrow_book()
book1.borrow_book()
book1.borrow_book()  

print("\nAfter Borrowing:")
book1.display()
```

#### 10. Product
#### Attributes
#### • product_name
#### • quantity
#### Method
#### sell(quantity)
#### Quantity should never become negative.

```python
class Product:
    
    def __init__(self, product_name, quantity):
        self.product_name = product_name
        self.quantity = quantity

    def sell(self, qty):
        if qty <= self.quantity:
            self.quantity -= qty
            print(qty, "product(s) sold successfully.")
        else:
            print("Insufficient quantity!")

    def display(self):
        print("Product Name:", self.product_name)
        print("Available Quantity:", self.quantity)


p1 = Product("Laptop", 10)

print("Before Selling:")
p1.display()

p1.sell(4)

p1.sell(8)

print("\nAfter Selling:")
p1.display()
```

#### 13. Fuel Tank
#### Attributes
#### • capacity
#### • current_fuel
#### Method
#### fill(amount)
#### Fuel should never exceed tank capacity.

```python
class FuelTank:
    
    def __init__(self, capacity, current_fuel):
        self.capacity = capacity
        self.current_fuel = current_fuel

    def fill(self, amount):
        if self.current_fuel + amount <= self.capacity:
            self.current_fuel += amount
            print("Fuel filled successfully.")
        else:
            print("Tank capacity exceeded!")

    def display(self):
        print("Tank Capacity:", self.capacity, "liters")
        print("Current Fuel:", self.current_fuel, "liters")


tank1 = FuelTank(50, 30)

print("Before Filling:")
tank1.display()

tank1.fill(15)

tank1.fill(10)

print("\nAfter Filling:")
tank1.display()
```

#### 15. Wallet
#### Attributes
#### • owner
#### • balance
#### Method
#### purchase(amount)
#### Purchase allowed only if enough money exists.

```python
class Wallet:
    
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def purchase(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Purchase successful.")
        else:
            print("Insufficient balance!")

    def display(self):
        print("Owner:", self.owner)
        print("Balance:", self.balance)


w1 = Wallet("Rahul", 5000)

print("Before Purchase:")
w1.display()

w1.purchase(1500)

w1.purchase(4000)

print("\nAfter Purchase:")
w1.display()
```

#### 16. Cricket Player
#### Attributes
#### • player_name
#### • total_runs
#### • matches
#### Method
#### add_runs(run)
#### Update total runs.
#### Print average after every update.
#### Average = total_runs / matches

```python
class CricketPlayer:
    def __init__(self, player_name, total_runs, matches):
        self.player_name = player_name
        self.total_runs = total_runs
        self.matches = matches

    def add_runs(self, run):
        self.total_runs += run
        average = self.total_runs / self.matches
        print("Total Runs:", self.total_runs)
        print("Average:", average)


# Create object
player = CricketPlayer("Virat", 5000, 100)

# Add runs
player.add_runs(80)
player.add_runs(50)
```

#### 17. School Fees
#### Attributes
#### • student_name
#### • total_fee
#### • paid_fee
#### Method
#### pay_fee(amount)
#### Fee paid cannot exceed total fee.

```python
class SchoolFees:
    def __init__(self, student_name, total_fee, paid_fee):
        self.student_name = student_name
        self.total_fee = total_fee
        self.paid_fee = paid_fee

    def pay_fee(self, amount):
        if self.paid_fee + amount <= self.total_fee:
            self.paid_fee += amount
            print("Fee Paid Successfully")
            print("Total Paid:", self.paid_fee)
            print("Remaining Fee:", self.total_fee - self.paid_fee)
        else:
            print("Payment exceeds total fee!")


# Create object
student = SchoolFees("Rahul", 50000, 20000)

# Pay fee
student.pay_fee(10000)
student.pay_fee(25000)
```

#### 19. ATM
#### Attributes
#### • balance
#### Methods
#### deposit()
#### withdraw()
#### withdraw()
#### should reject
#### • negative amount
#### • zero
#### • insufficient balance

```python
class ATM:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print("Deposit Successful")
            print("Balance:", self.balance)
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount")
        elif amount > self.balance:
            print("Insufficient Balance")
        else:
            self.balance -= amount
            print("Withdrawal Successful")
            print("Balance:", self.balance)


# Create object
atm = ATM(5000)

# Deposit money
atm.deposit(2000)

# Withdraw money
atm.withdraw(3000)
atm.withdraw(0)
atm.withdraw(-500)
atm.withdraw(5000)
```

#### 20. Hotel Room
#### Attributes
#### • room_number
#### • occupied
#### Method
#### book()
#### Room should not be booked twice.

```python
class HotelRoom:
    def __init__(self, room_number, occupied):
        self.room_number = room_number
        self.occupied = occupied

    def book(self):
        if self.occupied:
            print("Room is already booked")
        else:
            self.occupied = True
            print("Room booked successfully")


# Create object
room = HotelRoom(101, False)

# Book room
room.book()

# Try booking again
room.book()
```
