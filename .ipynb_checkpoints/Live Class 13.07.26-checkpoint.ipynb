# Nested Condition

```python
# Syntax :

if Condition :
   
    if condition :
            #block of code
    else : 
            #block of code :
else :
    #block of code :
```

```python
# Example :

Age = 18
ID = True

if Age >=18 : 
    if ID ==  True :
        print("Allow")
    else :
        print("Bring The ID")
else :
    print("Not Allow")
```

```python
student_marks = 60 

if student_marks >45:
  
    if student_marks >=80:
        print("Excellent mark")
    else :
          print("pass")
else : 
    print("Failed")
```

```python
A = 10
B = 20
C = 30

if A>B :
    if A>C :
        print("A Large")
    else :
        print("C large")
else :
    if B>C :
        print("B Large")
    else :
       print("C Large")
    
```
