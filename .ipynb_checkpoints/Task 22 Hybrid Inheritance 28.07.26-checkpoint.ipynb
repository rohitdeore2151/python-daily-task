# Hybrid Inheritance Questions (1-10)

#### 1. Create a hybrid inheritance structure with classes A → B and C → D (where D inherits from both B and C).
#### Concept
#### Hybrid Inheritance (Combination of Single + Multiple Inheritance)
#### What to do
#### • Create parent class A.
#### • Create class B inheriting from A.
#### • Create another class C.
#### • Create class D inheriting from both B and C.
#### How to solve
#### 1. Create class A.
#### 2. Add one method.
#### 3. Create class B(A).
#### 4. Create class C.
#### 5. Create class D(B, C).
#### 6. Create an object of D.
#### 7. Call methods from all parent classes.

```python
class A:
    def method_A(self):
        print("Method from class A")

class B(A):
    def method_B(self):
        print("Method from class B")

class C:
    def method_C(self):
        print("Method from class C")

class D(B, C):
    def method_D(self):
        print("Method from class D")

obj = D()

obj.method_A()
obj.method_B()
obj.method_C()
obj.method_D()
```

#### 2. Create Person, Employee, Student, and TeachingAssistant using hybrid inheritance.
###### Concept
###### Real-world Hybrid Inheritance
###### What to do
###### • Person is the base class.
###### • Employee inherits Person.
###### • Student inherits Person.
###### • TeachingAssistant inherits Employee and Student.
###### How to solve
###### Hierarchy
#### Person
###### / \
###### Employee Student
###### \ /
###### TeachingAssistant
###### Steps
###### 1. Store name in Person.
###### 2. Employee stores salary.
###### 3. Student stores marks.
###### 4. TeachingAssistant stores department.
###### 5. Display all information.

```python
class Person:
    def __init__(self, name):
        self.name = name

    def display_person(self):
        print("Name:", self.name)

class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name)
        self.salary = salary

    def display_employee(self):
        print("Salary:", self.salary)

class Student(Person):
    def __init__(self, name, marks):
        super().__init__(name)
        self.marks = marks

    def display_student(self):
        print("Marks:", self.marks)

class TeachingAssistant(Employee, Student):
    def __init__(self, name, salary, marks, department):
        Person.__init__(self, name)
        self.salary = salary
        self.marks = marks
        self.department = department

    def display_TA(self):
        print("Department:", self.department)

teacher = TeachingAssistant("Rohit", 40000, 85, "Computer Science")

teacher.display_person()
teacher.display_employee()
teacher.display_student()
teacher.display_TA()
```

#### 3. Create Vehicle → Car, Electric → ElectricCar (ElectricCar inherits from both).
###### Concept
###### Hybrid Inheritance
###### What to do
###### • Vehicle is parent.
###### • Car inherits Vehicle.
###### • Electric is another parent.
###### • ElectricCar inherits Car and Electric.
###### How to solve
###### Hierarchy
###### Vehicle
###### |
###### Car Electric
###### \ /
###### ElectricCar
###### Steps
###### 1. Vehicle has start().
###### 2. Car has drive().
###### 3. Electric has battery().
###### 4. ElectricCar accesses all methods.

```python
class Vehicle:
    def start(self):
        print("Starting Vehicale....")

class Car(Vehicle):
    def drive(self):
        print("Drive To Car.....")

class Electric:
    def bettery(self):
        print("Battery Is Charging....")

class ElectricCar(Car, Electric):
    def display(self):
        print("This is an Electric Car")

obj = ElectricCar()

obj.start()      
obj.drive()      
obj.bettery()    
obj.display()
```

#### 4. Demonstrate Hybrid Inheritance and print Method Resolution Order (MRO).
###### Concept
###### Method Resolution Order (MRO)
###### What to do
###### Create
###### A
###### |
###### B

###### C

###### D(B,C)
###### Print
###### D.mro()
###### How to solve
###### 1. Create four classes.
###### 2. Give same method name in multiple classes.
###### 3. Print ClassName.mro().
###### 4. Observe Python's searching order.

```python
class A:
    def show(self):
        print("class A")

class B(A):
    def show(self):
        print("Class B")

class C:
    def show(self):
        print("Class C")

class D(B,C):
    pass

obj = D()

obj.show()
```

#### 5. Create class A, B, C, D and solve the Diamond Problem using super().
###### Concept
###### Diamond Problem
###### Hierarchy
###### A
###### / \
###### B C
###### \ /
###### D
###### What to do
###### Use super() everywhere.
###### How to solve
###### 1. Create constructor in A.
###### 2. Call super() in B.
###### 3. Call super() in C.
###### 4. Call super() in D.
###### 5. Create object of D.
###### 6. Observe A constructor runs only once.

```python
class A:
    def __init__(self):
        print("Class A")

class B(A):
    def __init__(self):
        super().__init__()
        print("Class B ")

class C(A):
    def __init__(self):
        super().__init__()
        print("Class C")

class D(B,C):
    def __init__(self,):
        super().__init__()
        print("Class D")

obj = D()
 
```

#### 6. Create a banking system using hybrid inheritance.
###### Concept
###### Hybrid + Real-world Example
###### What to do
###### Account
###### |
###### Savings

###### Loan

###### SavingsLoanAccount

###### How to solve
###### 1. Account stores account number.
###### 2. Savings stores balance.
###### 3. Loan stores loan amount.
###### 4. SavingsLoanAccount inherits both.
###### 5. Print all details.

```python
class Account:
    def __init__(self, account_number, **kwargs):
        super().__init__(**kwargs)
        self.account_number = account_number


class Saving(Account):
    def __init__(self, balance, **kwargs):
        super().__init__(**kwargs)
        self.balance = balance


class Loan(Account):
    def __init__(self, loan_amount, **kwargs):
        super().__init__(**kwargs)
        self.loan_amount = loan_amount


class SavingLoanAccount(Saving, Loan):
    def __init__(self, account_number, balance, loan_amount):
        super().__init__(
            account_number=account_number,
            balance=balance,
            loan_amount=loan_amount
        )

    def display(self):
        print("Account Number :", self.account_number)
        print("Balance        :", self.balance)
        print("Loan Amount    :", self.loan_amount)


obj = SavingLoanAccount(9420930001, 10000, 1000)
obj.display()
```

#### 7. Create Shape, Color, Circle, and ColoredCircle.
##### Concept
##### Hybrid Inheritance
##### What to do
##### Shape
##### |
##### Circle

##### Color

##### ColoredCircle
##### How to solve
##### 1. Shape has area().
##### 2. Color has color().
##### 3. Circle inherits Shape.
##### 4. ColoredCircle inherits Circle and Color.
##### 5. Display area and color.

```python
import math

class Shape:
    def area(self, radius):
        return math.pi * radius * radius

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

class Color:
    def color(self, color_name):
        print("Color:", color_name)

class ColoredCircle(Circle, Color):
    def __init__(self, radius, color_name):
        Circle.__init__(self, radius)
        self.color_name = color_name

    def display(self):
        print("Area of Circle:", self.area(self.radius))
        self.color(self.color_name)

obj = ColoredCircle(7, "Red")
obj.display()
```

#### 8. Create constructors at every level.
##### Concept
##### Constructor Chaining
##### What to do
##### Every class has its own constructor.
##### How to solve
##### 1. Create constructor in every class.
##### 2. Use super() everywhere.
##### 3. Print constructor execution order.
##### 4. Observe execution using MRO.

```python
class A:
    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        print("Constructor of Class A")


class B(A):
    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        print("Constructor of Class B")


class C(A):
    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        print("Constructor of Class C")


class D(B, C):
    def __init__(self):
        super().__init__()
        print("Constructor of Class D")


obj = D()
```

#### 9. Override methods in intermediate classes.
###### Concept
###### Method Overriding
###### What to do
###### A
###### |
###### B

###### C

###### D(B,C)
###### All classes have
###### show()
###### How to solve
###### 1. Create same method.
###### 2. Override in B.
###### 3. Override in C.
###### 4. Call from D.
###### 5. Observe MRO.

```python
class A:
    def show(self):
        print("Show from Class A")


class B(A):
    def show(self):
        print("Show from Class B")


class C(A):
    def show(self):
        print("Show from Class C")


class D(B, C):
    pass


obj = D()

obj.show()
```

#### 10. Create a Device → Phone, Camera → SmartPhone hierarchy.
###### Concept
###### Hybrid Inheritance
###### What to do
###### Device
###### |
###### Phone

###### Camera

###### SmartPhone
###### How to solve
###### 1. Device has power_on().
###### 2. Phone has call().
###### 3. Camera has click().
###### 4. SmartPhone accesses all methods.

```python
class Device:
    def power_on(self):
        print("Device is powered on")


class Phone(Device):
    def call(self):
        print("Calling...")


class Camera:
    def click(self):
        print("Photo clicked")


class SmartPhone(Phone, Camera):
    def use(self):
        print("Using SmartPhone")


obj = SmartPhone()

obj.power_on()
obj.call()
obj.click()
obj.use()
```
