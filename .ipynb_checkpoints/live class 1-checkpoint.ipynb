```python
# dictionary Data types 

dict1 = {"Name":"Rohit","Age":10,"mob-no":123}
dict1.keys()
```

```python
# fromkeys() : Method

keys = ["name","age","mob-no"]
dict2 = dict.fromkeys(keys,10)
print(dict2)
```

```python
keys = ["a","b","c"]
dict2 = dict.fromkeys(keys,20)
print(dict2)
```

```python
dict2.keys()
```

```python
# Values () : Access the all value dict
dict2.values()
```

```python
# Get() : using key access the value
dict2.get("a")
```

```python
# items() : access the all items from the key

dict2.items()
```

```python
# pop() : remove the item from the dict using key name

dict4 = {"Car":"Thar","Model":2026,"color":"Black"}

dict4.pop("Model")
dict4
```

```python
# popitem() : 

dict4 = {"Car":"Thar","Model":2026,"color":"Black"}
dict4.popitem()
dict4
```

```python
# update() :

dict4 = {"Car":"Thar","Model":2026,"color":"Black"}
#dict4.update({"English":"20"})    #-----first way
#dict4["English"] = "99"    #-----second Way
dict4.update({"english":"55"})  #-----third way
dict4
```
