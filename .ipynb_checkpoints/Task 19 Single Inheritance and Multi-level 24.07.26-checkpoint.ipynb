## Single-Level Inheritance (1–15)

#### 1. Class Name: Person → Student
#### Problem
#### Create a parent class named Person and a child class named Student.
#### Explanation
#### The Person class should store:
#### • name
#### • age
#### The Student class should inherit from Person and add:
#### • marks
#### Create a method in Student to display the student's name, age, and marks.

```python
# Parent Class
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age


# Child Class
class Student(Person):

    def __init__(self, name, age, marks):
        # Call Parent Class Constructor
        super().__init__(name, age)
        self.marks = marks

    # Method to display student details
    def display(self):
        print("Student Details")
        print("---------------")
        print("Name  :", self.name)
        print("Age   :", self.age)
        print("Marks :", self.marks)


# Create Object
std = Student("Rohit", 22, 85)

# Display Details
std.display()
```

#### 2. Class Name: Vehicle → Car
#### Problem
#### Create a parent class named Vehicle.
#### Explanation
#### The Vehicle class should contain a method:
#### • start()
#### Create a child class Car that:
#### • Adds a brand attribute.
#### • Overrides the start() method to display a car-specific starting message.

```python
# Parent Class
class Vehicle:

    def start(self):
        print("Vehicle is starting...")


# Child Class
class Car(Vehicle):

    def __init__(self, brand):
        self.brand = brand

    # Overriding Parent Method
    def start(self):
        print(f"{self.brand} car is starting with a key...")


# Create Object
car1 = Car("Toyota")

# Call Overridden Method
car1.start()
```

#### 5. Class Name: Shape → Circle
#### Problem
#### Create a parent class named Shape.
#### Explanation
#### Create a method:
#### • area()
#### Create child class:
#### • Circle
#### Store:
#### • radius
#### Override area() to calculate the area of a circle.

```python
class Shape:

    def area(self):
        print("Area of Shape")


class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        print("Area of Circle:", 3.14 * self.radius * self.radius)


c = Circle(7)


c.area()
```

#### 6. Class Name: BankAccount → SavingsAccount
#### Problem
#### Create a parent class named BankAccount.
#### Explanation
#### Provide methods:
#### • deposit()
#### • withdraw()
#### Create subclass:
#### • SavingsAccount
#### Add:
#### • calculate_interest()

```python
# Parent Class
class BankAccount:

    def __init__(self, balance):
        self.Balance = balance

    # Deposit Method
    def Deposit(self, amount):
        self.Balance = self.Balance + amount
        print("Deposited:", amount)
        print("Balance:", self.Balance)

    # Withdraw Method
    def Withdraw(self, amount):
        if amount <= self.Balance:
            self.Balance = self.Balance - amount
            print("Withdrawn:", amount)
            print("Balance:", self.Balance)
        else:
            print("Insufficient Balance")


# Child Class
class SavingsAccount(BankAccount):

    # Calculate Interest Method
    def calculate_interest(self, rate):
        interest = (self.Balance * rate) / 100
        self.Balance = self.Balance + interest

        print("Interest:", interest)
        print("Balance after Interest:", self.Balance)


# Main Program
account = SavingsAccount(50000)

account.Deposit(2000)
account.Withdraw(3000)
account.calculate_interest(5)
```

# Multi-Level Inheritance (16–30)

```python
"""
16. Class Name:
Person → Employee → Manager
Problem
Create a three-level inheritance hierarchy.
Explanation
Person
Store:
• name
↓
Employee
Add:
• employee_id
↓
Manager
Add:
• department
Display all information.
"""
```

```python
class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def __init__(self, name, employee_id):
        super().__init__(name)
        self.employee_id = employee_id

class Manager(Employee):
    def __init__(self, name, employee_id, department):
        super().__init__(name, employee_id)
        self.department = department

    def display(self):
        print("Name:", self.name)
        print("Employee ID:", self.employee_id)
        print("Department:", self.department)

m = Manager("Rohit", 101, "IT")
m.display()
```

```python
class Animal:
    def eat(self):
        print("Eating")

class Mammal(Animal):
    def walk(self):
        print("Walking")

class Dog(Mammal):
    def bark(self):
        print("Barking")

d = Dog()
d.eat()
d.walk()
d.bark()
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

class Electronics(Product):
    def __init__(self, name, price, warranty):
        super().__init__(name, price)
        self.warranty = warranty

class Smartphone(Electronics):
    def __init__(self, name, price, warranty, ram, storage):
        super().__init__(name, price, warranty)
        self.ram = ram
        self.storage = storage

    def display(self):
        print("Name:", self.name)
        print("Price:", self.price)
        print("Warranty:", self.warranty)
        print("RAM:", self.ram)
        print("Storage:", self.storage)

s = Smartphone("Samsung", 60000, "2 Years", "8GB", "128GB")
s.display()
```

```python
"""
25. Class Name:
Device → Computer → Laptop
Problem
Create multi-level inheritance.
Explanation
Device
• brand
Computer
• processor
Laptop
• battery
• portability
Display all details.
"""
```

```python
class Device:
    def __init__(self, brand):
        self.brand = brand

class Computer(Device):
    def __init__(self, brand, processor):
        super().__init__(brand)
        self.processor = processor

class Laptop(Computer):
    def __init__(self, brand, processor, battery, portability):
        super().__init__(brand, processor)
        self.battery = battery
        self.portability = portability

    def display(self):
        print("Brand:", self.brand)
        print("Processor:", self.processor)
        print("Battery:", self.battery)
        print("Portability:", self.portability)

l = Laptop("Dell", "Intel i7", "6 Hours", "Portable")
l.display()
```
