# Basic Hierarchical Inheritance (1–12)

#### 1. Person → Student, Teacher
#### Question
#### Create a parent class Person with attributes name and age. Create two child classes Student
#### and Teacher that inherit from Person.
#### Concept
#### • Hierarchical Inheritance
#### • Parent attributes inherited by multiple child classes.
#### How to Solve
#### 1. Create a parent class Person.
#### 2. Add name and age using a constructor.
#### 3. Create Student(Person).
#### 4. Create Teacher(Person).
#### 5. Create objects of both classes.
#### 6. Show that both children can access name and age.

```python
# Parent class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age


# Child class 1
class Student(Person):
    def display_student(self):
        print("Student Name:", self.name)
        print("Student Age:", self.age)


# Child class 2
class Teacher(Person):
    def display_teacher(self):
        print("Teacher Name:", self.name)
        print("Teacher Age:", self.age)


# Creating objects
s1 = Student("Rohit", 22)
t1 = Teacher("Yash", 21)

# Display details
s1.display_student()
print()
t1.display_teacher()
```

#### 5. Employee → Manager, Developer
#### Question
#### Create class Employee with attributes emp_id and salary. Create subclasses Manager and
#### Developer.
#### Concept
#### • Attribute inheritance.
#### How to Solve
#### 1. Create parent constructor.
#### 2. Store emp_id and salary.
#### 3. Create two child classes.
#### 4. Show inherited attributes.

```python
# Parent class
class Employee:
    def __init__(self, emp_id, salary):
        self.emp_id = emp_id
        self.salary = salary


# Child class 1
class Manager(Employee):
    def __init__(self, emp_id, salary):
        super().__init__(emp_id, salary)

    def display(self):
        print("Manager Details")
        print("Employee ID:", self.emp_id)
        print("Employee Salary:", self.salary)


# Child class 2
class Developer(Employee):
    def __init__(self, emp_id, salary):
        super().__init__(emp_id, salary)

    def display(self):
        print("Developer Details")
        print("Employee ID:", self.emp_id)
        print("Employee Salary:", self.salary)


# Creating objects
manager = Manager("SUMIT123", 100000)
developer = Developer("RD234", 20000)

# Display details
manager.display()
print()
developer.display()
```

# Hierarchical Inheritance with Constructors & Methods (13–22)

#### 13. Person Constructor
##### Concept
##### Using super() to call the parent constructor.
###### How to Solve
##### 1. Create constructor in parent.
##### 2. Create child constructors.
##### 3. Use super().__init__().
##### 4. Add child-specific attribute.
##### 5. Display all data.

```python
# Parent class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Child class
class Student(Person):
    def __init__(self, name, age, roll_no):
        super().__init__(name, age)
        self.roll_no = roll_no

    def display(self):
        print("Student Details")
        print("Name    :", self.name)
        print("Age     :", self.age)
        print("Roll No :", self.roll_no)

# Child class
class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject = subject

    def display(self):
        print("Teacher Details")
        print("Name    :", self.name)
        print("Age     :", self.age)
        print("Subject :", self.subject)

# Creating objects
s = Student("Rohit", 22, 101)
t = Teacher("Yash", 21, "Mathematics")

# Display details
s.display()
print()
t.display()
```

#### 15. Vehicle Constructor
#### Concept
#### Constructor inheritance.
#### How to Solve
#### 1. Parent stores brand.
#### 2. Car adds doors.
#### 3. Bike adds engine_cc.
#### 4. Use super().

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

class Car(Vehicle):
    def __init__(self, brand, doors):
        super().__init__(brand)
        self.doors = doors

class Bike(Vehicle):
    def __init__(self, brand, engine_cc):
        super().__init__(brand)
        self.engine_cc = engine_cc

car = Car("Toyota", 4)
bike = Bike("Honda", 150)

print(car.brand, car.doors)
print(bike.brand, bike.engine_cc)
```

#### 16. Shape Area
#### Concept
#### Polymorphism using overriding.
#### How to Solve
#### 1. Parent constructor.
#### 2. Circle calculates πr2.
#### 3. Square calculates side2.
#### 4. Override area().

```python
class Shape:
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        return 3.14 * self.r * self.r

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side * self.side

c = Circle(5)
s = Square(4)

print(c.area())
print(s.area())
```

#### 18. BankAccount Interest
#### Concept
#### Different child behavior.

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

class Savings(BankAccount):
    def interest(self):
        return self.balance * 0.05

class Current(BankAccount):
    def interest(self):
        return self.balance * 0.02

s = Savings(10000)
c = Current(10000)

print(s.interest())
print(c.interest())
```

# Problem / Practical Questions (23–30)

#### 23. School Management
#### Concept
#### Real-world hierarchical inheritance.
#### How to Solve
#### 1. Parent → Person
#### 2. Common attributes (name, age)
#### 3. Student → marks
#### 4. Teacher → subject
#### 5. Staff → department
#### 6. Display information.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, age, marks):
        super().__init__(name, age)
        self.marks = marks

    def display(self):
        print(self.name, self.age, self.marks)

class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject = subject

    def display(self):
        print(self.name, self.age, self.subject)

class Staff(Person):
    def __init__(self, name, age, department):
        super().__init__(name, age)
        self.department = department

    def display(self):
        print(self.name, self.age, self.department)

Student("Rohit", 20, 90).display()
Teacher("Yash", 35, "Math").display()
Staff("Santosh", 40, "Admin").display()
```

#### 24. Bank System
#### Concept
#### Three children inherit one parent.
#### How to Solve
#### 1. Parent → Account
#### 2. Store balance.
#### 3. Savings → interest.
#### 4. Current → overdraft.
#### 5. FixedDeposit → maturity.
#### 6. Show each separately.

```python
class Account:
    def __init__(self, balance):
        self.balance = balance

class Savings(Account):
    def show(self):
        print("Interest:", self.balance * 0.05)

class Current(Account):
    def show(self):
        print("Overdraft: 5000")

class FixedDeposit(Account):
    def show(self):
        print("Maturity:", self.balance * 1.08)

Savings(10000).show()
Current(10000).show()
FixedDeposit(10000).show()
```

#### 27. Employee Salary
#### Concept
#### Different salary calculations.
#### How to Solve
#### 1. Parent stores common data.
#### 2. FullTime calculates monthly salary.
#### 3. PartTime calculates hourly salary.
#### 4. Override calculation method.

```python
class Employee:
    def __init__(self, name):
        self.name = name

    def salary(self):
        pass

class FullTime(Employee):
    def __init__(self, name, monthly):
        super().__init__(name)
        self.monthly = monthly

    def salary(self):
        return self.monthly

class PartTime(Employee):
    def __init__(self, name, hours, rate):
        super().__init__(name)
        self.hours = hours
        self.rate = rate

    def salary(self):
        return self.hours * self.rate

f = FullTime("Ram", 50000)
p = PartTime("Sam", 40, 500)

print(f.salary())
print(p.salary())
```

#### 29. Notification System
#### Concept
#### Method overriding.
#### How to Solve
#### 1. Parent → Notification
#### 2. Method send()
#### 3. Email overrides.
#### 4. SMS overrides.
#### 5. Push overrides.
#### 6. Call send() using each object.

```python
class Notification:
    def send(self):
        print("Sending Notification")

class Email(Notification):
    def send(self):
        print("Sending Email")

class SMS(Notification):
    def send(self):
        print("Sending SMS")

class Push(Notification):
    def send(self):
        print("Sending Push Notification")

Email().send()
SMS().send()
Push().send()
```

#### 30. Your Own Hierarchical Design
#### Concept
#### Complete hierarchical inheritance program.
#### How to Solve
