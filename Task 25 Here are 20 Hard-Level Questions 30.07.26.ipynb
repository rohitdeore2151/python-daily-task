#### 1. Take a list of numbers. Using loop and operators, separate even and odd numbers, then calculate sum of even and product of odd numbers without  using built-in sum() or prod().

```python
numbers = list(map(int, input("Enter numbers separated by space: ").split()))

even = []
odd = []

even_sum = 0
odd_product = 1

for num in numbers:
    if num % 2 == 0:
        even.append(num)
        even_sum += num
    else:
        odd.append(num)
        odd_product *= num

print("Even Numbers :", even)
print("Odd Numbers :", odd)
print("Sum of Even Numbers :", even_sum)

if len(odd) == 0:
    print("No Odd Numbers")
else:
    print("Product of Odd Numbers :", odd_product)
```

#### 2. Create a program using nested loops and break/continue that prints all prime numbers between two user-given numbers, and also counts how many primes were found.

```python
start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))

count = 0

print("Prime Numbers are:")

for num in range(start, end + 1):

    if num < 2:
        continue

    for i in range(2, num):
        if num % i == 0:
            break
    else:
        print(num)
        count += 1

print("Total Prime Numbers:", count)
```

#### 4. Create a class BankAccount with private __balance.
#### • Add deposit and withdraw with validation
#### • Maintain a class attribute total_accounts
#### • Add a class method to show total accounts
#### • Overload + operator to transfer money from one account to another

```python
class BankAccount:

    total_accounts = 0

    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance
        BankAccount.total_accounts += 1

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposit Successful")
        else:
            print("Invalid Amount")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawal Successful")
        else:
            print("Insufficient Balance")

    def get_balance(self):
        return self.__balance

    @classmethod
    def show_total_accounts(cls):
        print("Total Accounts:", cls.total_accounts)

    def __add__(self, other):

        amount = int(input("Enter amount to transfer: "))

        if amount <= self.__balance:
            self.__balance -= amount
            other.__balance += amount
            print("Transfer Successful")
        else:
            print("Insufficient Balance")

        return other


acc1 = BankAccount("Rahul", 5000)
acc2 = BankAccount("Amit", 3000)

acc1.deposit(1000)
acc1.withdraw(500)

acc1 + acc2

print("Rahul Balance:", acc1.get_balance())
print("Amit Balance:", acc2.get_balance())

BankAccount.show_total_accounts()
```

#### 5. Create abstract class Shape with abstract methods area() and perimeter(). Implement Circle and Rectangle using super(). Add a class attribute to count total shapes created.

```python
from abc import ABC, abstractmethod
import math


class Shape(ABC):

    total_shapes = 0

    def __init__(self):
        Shape.total_shapes += 1

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class Circle(Shape):

    def __init__(self, radius):
        super().__init__()
        self.radius = radius

    def area(self):
        return math.pi * self.radius * self.radius

    def perimeter(self):
        return 2 * math.pi * self.radius


class Rectangle(Shape):

    def __init__(self, length, width):
        super().__init__()
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


c = Circle(5)
r = Rectangle(10, 4)

print("Circle Area:", c.area())
print("Circle Perimeter:", c.perimeter())

print("Rectangle Area:", r.area())
print("Rectangle Perimeter:", r.perimeter())

print("Total Shapes:", Shape.total_shapes)
```

#### 9. Create a class ComplexNumber and overload +, -, *, and == operators completely.

```python
class ComplexNumber:

    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    def __add__(self, other):
        return ComplexNumber(
            self.real + other.real,
            self.imag + other.imag
        )

    def __sub__(self, other):
        return ComplexNumber(
            self.real - other.real,
            self.imag - other.imag
        )

    def __mul__(self, other):
        real = self.real * other.real - self.imag * other.imag
        imag = self.real * other.imag + self.imag * other.real
        return ComplexNumber(real, imag)

    def __eq__(self, other):
        return self.real == other.real and self.imag == other.imag

    def __str__(self):
        return f"{self.real}+{self.imag}i"


c1 = ComplexNumber(2, 3)
c2 = ComplexNumber(4, 5)

print("Addition:", c1 + c2)
print("Subtraction:", c1 - c2)
print("Multiplication:", c1 * c2)
print("Equal:", c1 == c2)
```

#### 10. Create a class Student with private marks list.
#### • Add setter with validation (each mark 0-100)
#### • Calculate average
#### • Overload > operator to compare two students based on average

```python
class Student:

    def __init__(self, name):
        self.name = name
        self.__marks = []

    def set_marks(self, marks):

        valid = True

        for mark in marks:
            if mark < 0 or mark > 100:
                valid = False

        if valid:
            self.__marks = marks
        else:
            print("Invalid Marks")

    def average(self):
        total = 0

        for mark in self.__marks:
            total += mark

        return total / len(self.__marks)

    def __gt__(self, other):
        return self.average() > other.average()


s1 = Student("Rahul")
s1.set_marks([80, 85, 90])

s2 = Student("Amit")
s2.set_marks([70, 75, 80])

print("Rahul Average:", s1.average())
print("Amit Average:", s2.average())

if s1 > s2:
    print("Rahul has higher average")
else:
    print("Amit has higher average")
```

#### 11. Take a number from user. Using while loop, reverse the number and check if it is a palindrome (without converting to string).

```python
num = int(input("Enter a number: "))

original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed Number:", reverse)

if original == reverse:
    print("Palindrome Number")
else:
    print("Not a Palindrome Number")
```

```python
from abc import ABC, abstractmethod

class Account(ABC):

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    @abstractmethod
    def calculate_interest(self):
        pass


class SavingsAccount(Account):

    def calculate_interest(self):
        interest = self.balance * 0.05
        print("Savings Interest:", interest)


class CurrentAccount(Account):

    def calculate_interest(self):
        interest = self.balance * 0.02
        print("Current Interest:", interest)


s = SavingsAccount("Rahul", 50000)
c = CurrentAccount("Amit", 50000)

s.calculate_interest()
c.calculate_interest()
```

#### 13. Create a class Matrix and overload + and * operators for matrix addition and multiplication (with proper size validation).

```python
class Matrix:

    def __init__(self, matrix):
        self.matrix = matrix

    def __add__(self, other):

        if len(self.matrix) != len(other.matrix) or len(self.matrix[0]) != len(other.matrix[0]):
            return "Matrix size must be same."

        result = []

        for i in range(len(self.matrix)):
            row = []
            for j in range(len(self.matrix[0])):
                row.append(self.matrix[i][j] + other.matrix[i][j])
            result.append(row)

        return result

    def __mul__(self, other):

        if len(self.matrix[0]) != len(other.matrix):
            return "Matrix multiplication not possible."

        result = []

        for i in range(len(self.matrix)):
            row = []

            for j in range(len(other.matrix[0])):
                total = 0

                for k in range(len(other.matrix)):
                    total += self.matrix[i][k] * other.matrix[k][j]

                row.append(total)

            result.append(row)

        return result


m1 = Matrix([[1, 2], [3, 4]])
m2 = Matrix([[5, 6], [7, 8]])

print("Addition:")
print(m1 + m2)

print("Multiplication:")
print(m1 * m2)
```

#### 15. Create a class Time (hours, minutes, seconds) and overload + operator to add two time objects with proper carry-over.

```python
class Time:

    def __init__(self, hours, minutes, seconds):
        self.hours = hours
        self.minutes = minutes
        self.seconds = seconds

    def __add__(self, other):

        sec = self.seconds + other.seconds
        minute = self.minutes + other.minutes
        hour = self.hours + other.hours

        if sec >= 60:
            minute += sec // 60
            sec = sec % 60

        if minute >= 60:
            hour += minute // 60
            minute = minute % 60

        return Time(hour, minute, sec)

    def display(self):
        print(f"{self.hours}:{self.minutes}:{self.seconds}")


t1 = Time(2, 45, 50)
t2 = Time(3, 30, 20)

t3 = t1 + t2

print("Result Time:")
t3.display()
```

#### 16. Create abstract class Payment with abstract method process(). Implement CreditCard and UPI. Use Duck Typing to process any payment object.

```python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def process(self):
        pass


class CreditCard(Payment):

    def process(self):
        print("Payment processed using Credit Card")


class UPI(Payment):

    def process(self):
        print("Payment processed using UPI")


def make_payment(payment):
    payment.process()


card = CreditCard()
upi = UPI()

make_payment(card)
make_payment(upi)
```

#### 17. Create a class Employee with private salary.
####  • Setter should not allow more than 20% increase at once
#### • Keep history of salary changes using class attribute
#### • Add static method to calculate tax

```python
class Employee:

    salary_history = []

    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary
        Employee.salary_history.append(salary)

    def get_salary(self):
        return self.__salary

    def set_salary(self, new_salary):

        limit = self.__salary * 1.20

        if new_salary <= limit:
            self.__salary = new_salary
            Employee.salary_history.append(new_salary)
            print("Salary Updated")
        else:
            print("Increase cannot exceed 20%")

    @staticmethod
    def calculate_tax(salary):
        return salary * 0.10


emp = Employee("Rahul", 50000)

print("Current Salary:", emp.get_salary())

emp.set_salary(55000)
emp.set_salary(70000)

print("Salary History:", Employee.salary_history)

print("Tax:", Employee.calculate_tax(emp.get_salary()))
```

#### 20. Create a complete system:
#### • Abstract class Vehicle
#### • Multiple inheritance with Electric and Fuel
#### • Operator overloading for comparing vehicles based on speed
#### • Duck Typing for start() method
#### • Private attributes + getter/setter with validation

```python
from abc import ABC, abstractmethod


class Vehicle(ABC):

    def __init__(self, speed):
        self.__speed = speed

    def get_speed(self):
        return self.__speed

    def set_speed(self, speed):
        if speed > 0:
            self.__speed = speed
        else:
            print("Invalid Speed")

    @abstractmethod
    def start(self):
        pass

    def __gt__(self, other):
        return self.__speed > other.__speed


class Electric:

    def battery(self):
        print("Battery Powered")


class Fuel:

    def fuel(self):
        print("Fuel Powered")


class Car(Vehicle, Electric, Fuel):

    def __init__(self, speed):
        super().__init__(speed)

    def start(self):
        print("Car Started")


def vehicle_start(vehicle):
    vehicle.start()


car1 = Car(120)
car2 = Car(100)

vehicle_start(car1)

car1.battery()
car1.fuel()

if car1 > car2:
    print("Car1 is Faster")
else:
    print("Car2 is Faster")
```
