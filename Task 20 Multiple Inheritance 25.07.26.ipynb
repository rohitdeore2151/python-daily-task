# Basic Multiple Inheritance (1-10)

#### 1. Create two parent classes Father and Mother. Create a child class Child that inherits from both.

```python
class Father:
    def father_info(self):
        print("Father Class")

class Mother:
    def mother_info(self):
        print("Mother Class")

class Child(Father, Mother):
    pass

c = Child()
c.father_info()
c.mother_info()
```

#### 2. Define class A with method show() and class B with method display(). Create class C that
#### inherits from both and calls both methods.

```python
class A:
    def show(self):
        print("Show method from A")

class B:
    def display(self):
        print("Display method from B")

class C(A, B):
    pass

obj = C()
obj.show()
obj.display()
```

#### 3. Create classes Teacher and Author. Create class Person that inherits from both and has attributes from both.

```python
class Teacher:
    def __init__(self):
        self.subject = "Python"

class Author:
    def __init__(self):
        self.book = "Programming"

class Person(Teacher, Author):
    def __init__(self):
        Teacher.__init__(self)
        Author.__init__(self)

p = Person()
print(p.subject)
print(p.book)
```

#### 6. Define two classes with same method name. Create a child class that inherits from both and resolve the method conflict.

```python
class A:
    def show(self):
        print("Show from A")

class B:
    def show(self):
        print("Show from B")

class C(A, B):
    def show(self):
        A.show(self)
        B.show(self)

obj = C()
obj.show()
```

#### 10. Demonstrate Method Resolution Order (MRO) using multiple inheritance with three classes.

```python
class A:
    def show(self):
        print("A")

class B(A):
    def show(self):
        print("B")

class C(A):
    def show(self):
        print("C")

class D(B, C):
    pass

obj = D()
obj.show()

print(D.mro())
```

# Multiple Inheritance with Static Methods (11-18)

#### 11. Create two parent classes with static methods. Create a child class that inherits from
#### both and calls the static methods.

```python
class Parent1:
    @staticmethod
    def method1():
        print("Static Method from Parent1")

class Parent2:
    @staticmethod
    def method2():
        print("Static Method from Parent2")

class Child(Parent1, Parent2):
    pass

Child.method1()
Child.method2()
```

#### 12. Define class MathUtils with static method add(). Create class AdvancedMath that
#### inherits and uses the static method.

```python
class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b

class AdvancedMath(MathUtils):
    pass

print(AdvancedMath.add(10, 20))
```

#### 13. Create class Validator with static method is_valid(). Create class Form that inherits and uses it.

```python
class Validator:
    @staticmethod
    def is_valid(text):
        return len(text) > 0

class Form(Validator):
    pass

print(Form.is_valid("Python"))
print(Form.is_valid(""))
```

#### 14. Define class Converter with static method to_celsius(). Create class WeatherApp that
#### inherits from multiple classes including Converter.

```python
class Converter:
    @staticmethod
    def to_celsius(f):
        return (f - 32) * 5 / 9

class Weather:
    pass

class WeatherApp(Converter, Weather):
    pass

print(WeatherApp.to_celsius(98.6))
```

#### 15. Create class Logger with static method log(). Create class Application that inherits from
#### Logger and another class.

```python
class Logger:
    @staticmethod
    def log(msg):
        print("LOG:", msg)

class Database:
    def connect(self):
        print("Database Connected")

class Application(Logger, Database):
    pass

app = Application()
app.connect()
Application.log("Application Started")
```

#### 17. Create class Calculator with static method and another class Display. Create child class that uses both.

# Multiple Inheritance with Abstract Methods (19-25)

#### 19. Create two abstract classes Shape and Color. Create a concrete class that inherits from
#### both and implements abstract methods.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def draw(self):
        pass

class Color(ABC):
    @abstractmethod
    def fill(self):
        pass

class Circle(Shape, Color):
    def draw(self):
        print("Drawing Circle")

    def fill(self):
        print("Filling Red Color")

obj = Circle()
obj.draw()
obj.fill()
```

#### 20. Define abstract class Payment and abstract class Notification. Create class OnlinePayment that inherits from both.

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self):
        pass

class Notification(ABC):
    @abstractmethod
    def notify(self):
        pass

class OnlinePayment(Payment, Notification):
    def pay(self):
        print("Payment Successful")

    def notify(self):
        print("Notification Sent")

obj = OnlinePayment()
obj.pay()
obj.notify()
```

#### 21. Create abstract class Vehicle and abstract class Electric. Create class ElectricCar that inherits from both.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Electric(ABC):
    @abstractmethod
    def charge(self):
        pass

class ElectricCar(Vehicle, Electric):
    def start(self):
        print("Electric Car Started")

    def charge(self):
        print("Battery Charging")

car = ElectricCar()
car.start()
car.charge()
```

#### 22. Define abstract class Readable and abstract class Writable. Create class File that implements both.

```python
from abc import ABC, abstractmethod

class Readable(ABC):
    @abstractmethod
    def read(self):
        pass

class Writable(ABC):
    @abstractmethod
    def write(self):
        pass

class File(Readable, Writable):
    def read(self):
        print("Reading File")

    def write(self):
        print("Writing File")

f = File()
f.read()
f.write()
```

#### 23. Create abstract class Flyable and abstract class Swimmable. Create class Duck that inherits from both.

```python
from abc import ABC, abstractmethod

class Flyable(ABC):
    @abstractmethod
    def fly(self):
        pass

class Swimmable(ABC):
    @abstractmethod
    def swim(self):
        pass

class Duck(Flyable, Swimmable):
    def fly(self):
        print("Duck is Flying")

    def swim(self):
        print("Duck is Swimming")

d = Duck()
d.fly()
d.swim()
```

#### 24. Demonstrate multiple inheritance where one parent is abstract and other is normal class.

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Pet:
    def owner(self):
        print("Pet Owner")

class Dog(Animal, Pet):
    def sound(self):
        print("Dog Barks")

d = Dog()
d.sound()
d.owner()
```

#### 25. Create two abstract classes with abstract methods of same name. Resolve in the child class.

```python
from abc import ABC, abstractmethod

class A(ABC):
    @abstractmethod
    def show(self):
        pass

class B(ABC):
    @abstractmethod
    def show(self):
        pass

class C(A, B):
    def show(self):
        print("Common Implementation for A and B")

obj = C()
obj.show()
```
