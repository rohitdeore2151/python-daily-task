### 30 Questions on List Comprehensions

#### Basic List Comprehensions (1-10)

```python
# 1. Write a list comprehension to create a list of numbers from 1 to 10.

number = [print(i) for i in range(1,11)]
```

```python
# 2. Create a list of squares of numbers from 1 to 10 using list comprehension.

list1 = []
number = [list1.append(i**2) for i in range(1,11)]
print(list1)
```

```python
# 3. Write a list comprehension to get even numbers from 1 to 20.

even_numbers = [x for x in range(1, 21) if x % 2 == 0]
print(even_numbers)
```

```python
# 4. Convert the following loop into a list comprehension:
# result = []
# for i in range(10):
# result.append(i*3)

result = []
for i in range(10):
    result.append(i * 3)
print(result)
```

```python
# 5. What is the output of [x for x in range(5)]?

print([x for x in range(5)])
```

```python
# 7. Write a list comprehension to convert all strings in a list to uppercase.

words = ["apple", "banana", "cherry"]
uppercase_words = [word.upper() for word in words]

print(uppercase_words)
```

```python
# 8. Predict the output: [i**2 for i in range(1,6)]

result = [] 
for i in range(1,6):
    result.append(i**2)
print(result)
```

```python
# 9. Create a list containing numbers from 0 to 9 using list comprehension.

lists = [i for i in range(10)]
print(lists)
```

```python
# 10.Write a list comprehension to get only positive numbers from [-2, -1, 0, 1, 2,3].

numbers = [-2, -1, 0, 1, 2, 3]
positive_numbers = [x for x in numbers if x > 0]

print(positive_numbers)
```

### List Comprehension with Conditions (11-20)

```python
# 11.Write a list comprehension to get numbers divisible by 3 between 1 and 30.

divisible_by_3 = [x for x in range(1, 31) if x % 3 == 0]
print(divisible_by_3)
```

```python
# 12.Filter all odd numbers from 1 to 15 using list comprehension.

odd_numbers = [x for x in range(1, 16) if x % 2 != 0]
print(odd_numbers)
```

```python
# 13.Use if-else in list comprehension to label numbers as "Even" or "Odd" for 1 to 8.

labels = ["Even" if x % 2 == 0 else "Odd" for x in range(1, 9)]
print(labels)
```

```python
# 14.Create a list of squares of only even numbers from 1 to 10.

squares = [x**2 for x in range(1, 11) if x % 2 == 0]
print(squares)
```

```python
# 15.What will be the output: [x for x in range(10) if x % 2 == 0]

[x for x in range(10) if x % 2 == 0]
```

```python
# 16.Write a list comprehension to remove vowels from a string "Python Programming".

text = "Python Programming"

result = ''.join([ch for ch in text if ch.lower() not in "aeiou"])

print(result)
```

```python
# 17.Get all words with length greater than 4 from a list of words.

words = ["Python", "is", "a", "powerful", "programming", "language"]

result = [word for word in words if len(word) > 4]

print(result)
```

```python
# 18.Use list comprehension to create a list of tuples: [(i, i**2) for i in range(5)]

result = [(i, i**2) for i in range(5)]

print(result)
```

```python
# 19.Write a list comprehension to replace all negative numbers with 0 in a list.

numbers = [5, -3, 8, -1, 0, -7, 4]

result = [0 if num < 0 else num for num in numbers]

print(result)
```
