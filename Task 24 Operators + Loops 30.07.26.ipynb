#### 14. Create class BankAccount with private attribute __balance. Add deposit and withdraw methods with validation.

```python
class BankAccount:
    def __init__(self):
        self.__balance = 0

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient Balance")

    def show(self):
        print(self.__balance)

b = BankAccount()

b.deposit(5000)
b.withdraw(2000)
b.show()
```

#### 17. Create class Product with private __price. Add getter and setter with condition that price cannot be negative.

#### 18. Create class Rectangle with private length and width. Add getter, setter and method to calculate area.

```python
class Rectangle:
    def __init__(self, length, width):
        self.__length = length
        self.__width = width

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def set_length(self, length):
        self.__length = length

    def set_width(self, width):
        self.__width = width

    def area(self):
        return self.__length * self.__width

r = Rectangle(10, 5)

print(r.area())
```

## Inheritance + Abstraction + super() (19-24)

#### 19. Create single level inheritance: Person → Student. Add attributes in both classes.

```python
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, marks):
        super().__init__(name)
        self.marks = marks

s = Student("Yash", 90)

print(s.name)
print(s.marks)
```

#### 20. Create multi-level inheritance: Person → Employee → Manager.

```python
class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name)
        self.salary = salary

class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department

m = Manager("Rohit", 50000, "IT")

print(m.name)
print(m.salary)
print(m.department)
```

#### 21. Create hierarchical inheritance: Parent Animal → Children Dog and Cat.

```python
class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Barking")

class Cat(Animal):
    def meow(self):
        print("Meowing")

d = Dog()
c = Cat()

d.eat()
d.bark()

c.eat()
c.meow()
```

#### 23. Create abstract class Shape with abstract method area(). Implement it in Circle and Rectangle using super().

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    def __init__(self):
        super().__init__()

    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        super().__init__()
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

class Rectangle(Shape):
    def __init__(self, length, width):
        super().__init__()
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

c = Circle(5)
r = Rectangle(10, 4)

print(c.area())
print(r.area())
```

#### 24. Create hybrid inheritance with classes A, B, C and D (D inherits from B and C).

```python
class A:
    def showA(self):
        print("A")

class B(A):
    def showB(self):
        print("B")

class C(A):
    def showC(self):
        print("C")

class D(B, C):
    def showD(self):
        print("D")

d = D()

d.showA()
d.showB()
d.showC()
d.showD()
```

## Method Overloading, Overriding, Operator Overloading + Duck Typing (25-30)

#### 25. Create a class Calculator that can add 2 or 3 numbers using method overloading concept (*args or default arguments).

#### 26. Create parent class Animal with method sound(). Override it in Dog and Cat.

#### 27. Create class Point and overload + operator to add two points.

```python

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def display(self):
        print(self.x, self.y)

p1 = Point(2, 3)
p2 = Point(4, 5)

p3 = p1 + p2
p3.display()
```

#### 28. Create class Book and overload > operator to compare number of pages of two books.

```python
class Book:
    def __init__(self, pages):
        self.pages = pages

    def __gt__(self, other):
        return self.pages > other.pages

b1 = Book(250)
b2 = Book(300)

print(b1 > b2)
```

#### 29. Create two different classes Duck and Person both having method walk(). Demonstrate Duck Typing.

```python
class Duck:
    def walk(self):
        print("Duck is walking")

class Person:
    def walk(self):
        print("Person is walking")

def move(obj):
    obj.walk()

d = Duck()
p = Person()

move(d)
move(p)
```

#### 30. Create class ComplexNumber and overload both + and - operators.

```python
class ComplexNumber:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    def __add__(self, other):
        return ComplexNumber(self.real + other.real, self.imag + other.imag)

    def __sub__(self, other):
        return ComplexNumber(self.real - other.real, self.imag - other.imag)

    def display(self):
        print(self.real, "+", self.imag, "i")

c1 = ComplexNumber(5, 3)
c2 = ComplexNumber(2, 1)

c3 = c1 + c2
c4 = c1 - c2

c3.display()
c4.display()
```
