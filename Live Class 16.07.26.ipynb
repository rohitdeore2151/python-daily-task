```python
# what is list comprehension:  combined[ loop+Condition+Expression] inside one list

# Syntax: [expression for item in place condition]  output means expression

list1 = [1,2,3,4,5,6,7]

result = ["Even"for num in list1 if num % 2 == 0]

list(result)
```

```python
# if condition   :

# [ expression for item in place condition ]

#  else condition:

# [ expresiton if condition else expression for  item]

list1 = [1,2,3,4,5,6,7]

# using if condition 

result_even = ["Even" for num in list1 if num % 2 == 0]

# using if else

result_even_odd["even" if num % 2 == 0 else "odd:,{num}" for num in list1 ]

print(f"using if answer :{result_even}")
print(f"using if else answer : {result_even_odd}")
```
