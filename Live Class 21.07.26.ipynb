```python
# bank management :

class Bank_Management:
    #class Attribute
    bank_name = "State Bank Of India"
    # create a attribute
    def __init__(self,ac_no,balance,branch):
        self.Account_Number = ac_no
        self.Balance = balance
        self.Branch = branch

    # display all attribute
    # method 1:

    def display_bank_details(self):
        print(f"Account Number:{self.Account_Number}")
        print(f"Balance:{self.Balance}")
        print(f"Branch:{self.Branch}")
        print(f"Bank_Name: {Bank_Management.bank_name}")

    # Method 2

    def deposit_money(self,amt):
        self.balance +=amt
        print(f"Deposite {self.Balance}Successfully Dne !")

    #method 3:
    

    def withdraw_money(self,amt):
        self.Balance -=amt
        print(f"Withdraw Money {amt}Successfully Done !")

# create the class object

bank = Bank_Management(12345,15000,"Dhule")  

# call the withdraw method:

bank.display_bank_details()
```
