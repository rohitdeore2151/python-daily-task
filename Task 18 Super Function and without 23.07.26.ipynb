#### 1. Class Name: Shape
#### Problem
#### Create an abstract class named Shape.
#### Explanation
#### A shape can have different formulas for area and perimeter. Therefore, create two abstract
#### methods:
#### • area()
#### • perimeter()
#### Create subclasses like Circle and Rectangle and implement both methods.

```python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

    def perimeter(self):
        return 2 * 3.14 * self.radius


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


c = Circle(5)
print("Circle Area:", c.area())
print("Circle Perimeter:", c.perimeter())

r = Rectangle(10, 5)
print("Rectangle Area:", r.area())
print("Rectangle Perimeter:", r.perimeter())
```

#### 15. Class Name: Shape3D
#### Problem
#### Create an abstract class named Shape3D.
#### Explanation
#### Three-dimensional shapes have different formulas.
#### Create abstract methods:
#### • volume()
#### • surface_area()
#### Create subclasses:
#### • Cube
#### • Sphere

```python
from abc import ABC, abstractmethod

# Abstract class
class Shape3D(ABC):

    @abstractmethod
    def volume(self):
        pass


# Cube class
class Cube(Shape3D):

    def __init__(self, side):
        self.side = side

    def volume(self):
        return self.side ** 3


# Sphere class
class Sphere(Shape3D):

    def __init__(self, radius):
        self.radius = radius

    def volume(self):
        return (4/3) * 3.14 * (self.radius ** 3)


# Creating objects
cube = Cube(5)
sphere = Sphere(3)

# Calling methods
print("Cube Volume:", cube.volume())
print("Sphere Volume:", sphere.volume())
```

#### __init__(self, color)

#### • area()

#### • Circle
#### Use super() to initialize the color.

```python
from abc import ABC, abstractmethod

# Abstract class
class Shape(ABC):

    def __init__(self, color):
        self.color = color

    @abstractmethod
    def area(self):
        pass


# Circle class
class Circle(Shape):

    def __init__(self, color, radius):
        super().__init__(color)   # Initialize color from parent class
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2


# Creating object
circle = Circle("Red", 5)

# Calling methods
print("Shape Color:", circle.color)
print("Circle Area:", circle.area())
```

#### • Vehicle → Car → ElectricCar

```python
from abc import ABC, abstractmethod

# Abstract class
class Vehicle(ABC):
    def __init__(self, brand):
        self.brand = brand

    @abstractmethod
    def display(self):
        pass


# Child class
class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)   # Initialize Vehicle data
        self.model = model

    def display(self):
        print("Brand:", self.brand)
        print("Model:", self.model)


# Grandchild class
class ElectricCar(Car):
    def __init__(self, brand, model, battery):
        super().__init__(brand, model)   # Initialize Car data
        self.battery = battery

    def display(self):
        super().display()
        print("Battery:", self.battery)


# Create object
car = ElectricCar("Tesla", "Model 3", "75 kWh")

car.display()
```

#### 30. Class Name: Report
#### Problem
#### Create an abstract class named Report.
#### Explanation
#### Store:
#### • Report title
#### Add a normal method:
#### • show_header()
#### Create abstract method:
#### • generate()
#### Create subclasses:
#### • PDFReport
#### • HTMLReport
#### Use super() to:
#### • initialize the title using super().__init__()
#### • call the parent method using super().show_header()
#### • then generate the report in the subclass..

```python
from abc import ABC, abstractmethod

# Abstract class
class Report(ABC):
    def __init__(self, title):
        self.title = title

    def show_header(self):
        print("Report Title:", self.title)

    @abstractmethod
    def generate(self):
        pass


# Subclass 1
class PDFReport(Report):
    def __init__(self, title):
        super().__init__(title)   # Initialize title

    def generate(self):
        super().show_header()      # Call parent method
        print("Generating PDF Report")


# Subclass 2
class HTMLReport(Report):
    def __init__(self, title):
        super().__init__(title)   # Initialize title

    def generate(self):
        super().show_header()      # Call parent method
        print("Generating HTML Report")


# Create objects
pdf = PDFReport("Monthly Sales Report")
html = HTMLReport("Website Report")

pdf.generate()
html.generate()
```
