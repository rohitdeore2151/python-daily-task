## Problem 6

#### Create a custom iterator that prints numbers from 1 to 10.

```python
class NumberIterator:
    def __init__(self):
        self.current = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= 10:
            number = self.current
            self.current += 1
            return number
        else:
            raise StopIteration

numbers = NumberIterator()

for num in numbers:
    print(num)
```

## Problem 8

#### Create a custom iterator that prints numbers in reverse order from 10 to 1.

```python
class ReverseIterator:
    def __init__(self):
        self.current = 10

    def __iter__(self):
        return self

    def __next__(self):
        if self.current >= 1:
            number = self.current
            self.current -= 1
            return number
        else:
            raise StopIteration

reverse_numbers = ReverseIterator()

for num in reverse_numbers:
    print(num)
```

## Problem 9

#### Create a custom iterator that prints the characters of the word "GENERATOR" one by one.

```python
class WordIterator:
    def __init__(self, word):
        self.word = word
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.word):
            char = self.word[self.index]
            self.index += 1
            return char
        else:
            raise StopIteration

letters = WordIterator("GENERATOR")


for ch in letters:
    print(ch)
```

## Problem 16

#### Given the following list:
#### numbers = [10, 15, 20, 25, 30, 35]
#### Create a generator that yields only the even numbers.

```python
numbers = [ 10,15,20,25,30,35]

def even_numbers(nums):
    for num in nums:
        if num % 2 == 0:
            yield num
            
for n in even_numbers(numbers):
    print(n)
```

## Problem 24
#### Create a custom iterator that prints all vowels from the string below.
#### text = "Artificial Intelligence"

```python
class VowelIterator:
    def __init__(self, text):
        self.text = text
        self.index = 0
        self.vowels = "AEIOUaeiou"

    def __iter__(self):
        return self

    def __next__(self):
        while self.index < len(self.text):
            char = self.text[self.index]
            self.index += 1
            if char in self.vowels:
                return char
        raise StopIteration

# Given string
text = "Artificial Intelligence"

# Create iterator object
vowels = VowelIterator(text)

# Print vowels
for v in vowels:
    print(v)
```

## Problem 25

#### Create a custom iterator that prints the Fibonacci sequence up to 10 terms.

```python
class Fibonacci:
    def __init__(self):
        self.a = 0
        self.b = 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < 10:
            value = self.a
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return value
        else:
            raise StopIteration

fib = Fibonacci()

for num in fib:
    print(num)
```

## Problem 26

#### Create a generator that yields the Fibonacci sequence up to 15 terms.

```python
def fibonacci():
    a = 0
    b = 1

    for i in range(15):
        yield a
        a, b = b, a + b

generator = fibonacci()

for num in generator:
    print(num)
```

## Problem 27

#### Create a generator that yields all prime numbers between 1 and 100.

```python
def prime_numbers():
    for num in range(2, 101):
        is_prime = True

        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            yield num

generator = prime_numbers()

for prime in generator:
    print(prime)
```

## Problem 28

#### Create a generator that yields all factors of a given number.
#### Example Input:
#### num = 24

```python
def factors(n):
    for i in range(1, n + 1):
        if n % i == 0:
            yield i


num = 24

print("Factors:")
for factor in factors(num):
    print(factor)
```

## Problem 30

#### Create an infinite generator that yields multiples of 5.
#### Print only the first 15 values using next().

```python
def multiples_of_5():
    num = 5
    while True:
        yield num
        num += 5

gen = multiples_of_5()

for _ in range(15):
    print(next(gen))
```
