```python
# check the even or odd :

def check_even_odd(num):
    if num % 2 == 0:
        return "even"
    else:
        return "odd"
check_even_odd(7)        
```

```python
# check the positive or negative Value

def check_positive_negative(num):
    if num >= 0:
        return "positive"
    else:
        return "negative"
check_positive_negative(-4)
```

```python
# Types Of Parameter:

# 1. No parameter :

def Hello():
    return"Good Morning"

Hello()
```

```python
# 2. Positional Parameters :

def info(name,age):
    return f"my name is {name} and age is {age}"

info("Rohit",22)
```

```python
# 3. Default Parameters :

def hello(name = "Rohit"):
    return f"hii {name}, Good Morning"

hello("Nayan")
```

```python
# 4. Keyword Arguments :

def info(name,age,mob_no):
    return f"My Nmae is {name},and Age {age},This My Mobile No {mob_no}"

info(name = "Rohit",age = 22, mob_no = 876753)
```

```python
#take the name , address,city,course,mentor_name:

def info(name,address,city,course,mentor_name):
    print(f"Name: {name} \nAddress: {address} \nCity :{city} \nCourse: {course} \nMentor Name: {mentor_name}")

info(name = "Rohit", address = "Dhule", city = "Borkund", course ="Data Scince" , mentor_name = "Sumit sir")
```

```python
# 5. *args (Multiple Position):

def add(*numbers):
    c = sum(numbers)
    return c
add(1,2,3,44,55,)
```

```python
# 6. **kwargs ( multiple keyword Argument):

def items(**details):
   for key,value in details.items(): 
       return f"key {key} : value {value}"
    
print(name = "Rohit", address = "Dhule", city = "Borkund", course ="Data Scince" , mentor_name = "Sumit sir")
```
