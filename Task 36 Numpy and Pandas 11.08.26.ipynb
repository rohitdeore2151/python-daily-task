#### 11.Clean the customer_segment column by standardizing capitalization and whitespace.

```python
import numpy as np

df["customer_segment"] = (
    df["customer_segment"]
    .str.strip()
    .str.replace(".", "", regex=False)
    .str.title()
    .replace("Unknoun", "Unknown")
)

df["customer_segment"].value_counts()
```

#### 12.Find all unique values in the gender column and convert them into consistent categories such as Male and Female.

```python
df["gender"].unique()
```

```python
df["gender"] = (
    df["gender"]
    .str.strip()
    .str.lower()
    .replace({
        "m": "Male",
        "male": "Male",
        "f": "Female",
        "female": "Female"
    })
)

df["gender"].value_counts()
```

#### 13.Clean the is_active column so values such as Yes, yes, Y, 1 become 1, and No, no, N, 0 become 0.

```python
df["is_active"] = (
    df["is_active"]
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({
        "yes": 1,
        "y": 1,
        "1": 1,
        "no": 0,
        "n": 0,
        "0": 0
    })
)

df["is_active"].value_counts()
```

#### 14.Convert the churn_flag column from categorical values (Yes/No) into a binary numerical column (1/0).

```python
df["churn_flag"] = (
    df["churn_flag"]
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({
        "yes": 1,
        "no": 0
    })
)

df["churn_flag"].value_counts()
```
