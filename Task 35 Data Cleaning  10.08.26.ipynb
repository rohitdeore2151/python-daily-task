#### 26.Split the Route column using → as the separator and create Route_1, Route_2, Route_3, Route_4, and Route_5.

#### 28.Replace the missing values in the route columns with an appropriate value such as None.

```python
route = df["Route"].str.split("→", expand=True)

route.columns = [f"Route_{i+1}" for i in range(route.shape[1])]

df = pd.concat([df, route], axis=1)


df[["Route_3", "Route_4", "Route_5"]] = df[
    ["Route_3", "Route_4", "Route_5"]
].fillna("None")


df.head()
```

#### 29.Drop the original Route column after creating and cleaning the route features.

```python
route = df["Route"].str.split("→", expand=True)

route.columns = [f"Route_{i+1}" for i in range(route.shape[1])]

df = pd.concat([df, route], axis=1)

df = df.fillna({"Route_3": "None", "Route_4": "None", "Route_5": "None"})

df.drop("Route", axis=1, inplace=True)

df.head()
```
