```python
# Sets Methods :

# add method : add element in the set
#syntax : set_name.add(element)

set = {1,2,3,4,5,6}
set.add(10)
set
```

```python
# Fast Membership test : 
# in :serching the value

sets = {1,2,3,4,5,6}

print(2 in sets)
```

```python
# update the value 
# syntax : set_name.update([element])

sets = {1,2,3,4,5,6}
sets.update([6,7])
sets
```

```python
# remove Method : remove the specific elements

sets = {1,2,3,4,5,6}
sets.remove(2)
sets
```

```python
# discard() : removes element from set (no error not found)

sets = {1,2,3,4,5,6}
sets.discard(5)
sets
```

```python
# pop Method : remove the random element 

sets = {-1,1,2,3,4,5,6}
sets.pop()             # first Integer value Remove
sets                   # largest negative value also remove 
```

```python
# clear Method () : remove all element to set 

sets = {-1,1,2,3,4,5,6}
sets.clear()
sets
```

```python
# copy Method () : 

sets1 = {1,2,3,4,5,6}
sets = sets1.copy()
sets
```

```python
# union :

A = {1,2,3,4,5,6,}

B = {7,8,9,0,2,4}
A.union(B)
```

```python
# intersection : select the comon value

A = {1,2,3,4,5,6,}

B = {7,8,9,0,2,4}

A.intersection(B)
```

```python
# difference () : chech the element not in B

A = {1,2,3,4,5,6,}

B = {7,8,9,0,2,4}

A.difference(B)
```
