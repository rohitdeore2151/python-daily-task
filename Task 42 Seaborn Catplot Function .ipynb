```python
import seaborn as sns
import matplotlib.pyplot as plt

titanic = sns.load_dataset("titanic")
tips = sns.load_dataset("tips")
```

#### 1. Survival by Gender
#### Use catplot() to show the number of passengers who survived and did not survive for each
#### gender.
#### • X-axis → sex
#### • Hue → survived
#### • Chart → count

```python
sns.catplot(
    data=titanic,
    x="sex",
    hue="survived",
    kind="count"
)

plt.title("Survival Count by Gender")
plt.show()
```

#### 2. Survival by Passenger Class
#### Create a categorical plot showing the number of passengers in each class.
#### Use:
#### • x="class"
#### • kind="count"

```python
sns.catplot(
    data=titanic,
    x="class",
    kind="count"
)

plt.title("Number of Passengers by Class")
plt.show()
```

#### 3. Survival Count by Class

#### 4. Survival Percentage by Gender
#### Create a categorical visualization comparing survival across sex.
#### Use catplot() and choose an appropriate categorical plot.
#### Then identify which gender has a higher survival rate.

```python
sns.catplot(
    data=titanic,
    x="sex",
    y="survived",
    kind="point",
    errorbar=None
)

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")
plt.show()
```

#### 7. Fare by Passenger Class
#### Use catplot() to compare fare across the three passenger classes.
#### Add:
#### • Appropriate color
#### • Title
#### • X/Y labels

```python
sns.catplot(
    data=titanic,
    x="class",
    y="fare",
    kind="box",
    hue = "class",
    palette="Set2",
    height=5,
    aspect=1.2
)

plt.title("Fare Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.show()
```

#### 13. Survival vs Age Group
#### Create an age_group column:
#### • Child → Age < 18
#### • Adult → 18–59
#### • Senior → 60+
#### Use catplot() to compare survival across these groups.

```python
import pandas as pd 

sns.catplot(
    data=titanic,
    x="age_group",
    y="survived",
    kind="bar",
    errorbar=None,
    order=["Child", "Adult", "Senior"]
)

plt.title("Survival Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Survival Rate")
plt.show()
```

#### 15. Titanic Multi-Panel Analysis
#### Create a catplot() with:
#### • X → class
#### • Y → fare
#### • Hue → sex
#### • Columns → survived
#### Use an appropriate categorical kind.
#### Add:
#### • Figure size
#### • Title
#### • Color palette
#### • Appropriate labels
#### Explain what you observe.

```python
g = sns.catplot(
    data=titanic,
    x="class",
    y="fare",
    hue="sex",
    col="survived",
    kind="box",
    palette="Set2",
    height=5,
    aspect=1.1
)

g.set_axis_labels("Passenger Class", "Fare")
g.set_titles("Survived = {col_name}")
g.fig.suptitle(
    "Fare by Class, Gender and Survival",
    y=1.05
)

plt.show()
```
