```python
num = 0 

even_odd = []

while num <= 10 :

    if num % 2 == 0:
        even_odd.append("Even")
    else:
        even_odd.append("Odd")
    num +=1

print(even_odd)
                        
    
```

```python
num = 1

dicts ={}

while num <=10:
    if num > 0:
        dicts[num]=("Positive")
    else:
         dicts[num]=("Negative")
    num +=1
print(dicts)
```

```python
num = 12345

rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10
print(rev)
```

```python
num = 12345

original = num 

rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

    if original == rev:
        result = "polidrom"
    else:
        result = "Not polidrom"
print(result)
```

```python
num = 1

result = []

while num <=10:
    if num % 2 == 0:
        result.append(f"even{num}:{num**2}")
    else:
        result.append(f"odd{num}:{num**3}")
    num +=1
print(result)
```
