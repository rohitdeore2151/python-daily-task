#### 11. Create class BankAccount with private __balance. Add deposit and withdraw methods. Raise custom exception if balance becomes negative.

```python
class NegativeBalanceError(Exception):
    pass

class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amt):
        self.__balance += amt

    def withdraw(self, amt):
        if self.__balance - amt < 0:
            raise NegativeBalanceError("Negative Balance")
        self.__balance -= amt

    def show(self):
        print("Balance =", self.__balance)

b = BankAccount(1000)
b.deposit(500)
b.withdraw(300)
b.show()
```

#### 14. Create class Rectangle with private length and width. Add method to calculate area and perimeter.

```python
class Rectangle:
    def __init__(self, l, w):
        self.__length = l
        self.__width = w

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

r = Rectangle(5, 4)
print(r.area())
print(r.perimeter())
```

#### 18. Create abstract class Shape with abstract method area(). Implement it in Circle and Square using super().

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    def __init__(self, value):
        self.value = value

    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, r):
        super().__init__(r)

    def area(self):
        return 3.14 * self.value * self.value

class Square(Shape):
    def __init__(self, s):
        super().__init__(s)

    def area(self):
        return self.value * self.value

print(Circle(5).area())
print(Square(4).area())
```

#### 21. Create class Point and overload + and > operators.

```python
class Point:
    def __init__(self, x):
        self.x = x

    def __add__(self, other):
        return Point(self.x + other.x)

    def __gt__(self, other):
        return self.x > other.x

p1 = Point(10)
p2 = Point(20)

p3 = p1 + p2
print(p3.x)
print(p2 > p1)
```

#### 26. Create a class Student with private marks list.
#### • Setter should validate each mark (0-100)
#### • Overload > operator to compare two students based on average
#### • Raise custom exception if marks list is empty

```python
class EmptyMarksError(Exception):
    pass

class Student:
    def __init__(self):
        self.__marks = []

    def set_marks(self, marks):
        for i in marks:
            if 0 <= i <= 100:
                self.__marks.append(i)

    def avg(self):
        if len(self.__marks) == 0:
            raise EmptyMarksError("Marks Empty")
        return sum(self.__marks)/len(self.__marks)

    def __gt__(self, other):
        return self.avg() > other.avg()

s1 = Student()
s2 = Student()

s1.set_marks([80,90])
s2.set_marks([70,75])

print(s1 > s2)
```

#### 29. Create a class Employee with private salary.
#### • Setter should not allow more than 15% increase at once
#### • Keep history of all salary changes
#### • Add static method to calculate tax
#### • Overload > operator to compare employees by salary

```python
class Employee:
    def __init__(self, sal):
        self.__salary = sal
        self.history = [sal]

    def set_salary(self, s):
        if s <= self.__salary * 1.15:
            self.__salary = s
            self.history.append(s)
        else:
            print("Increase >15% not allowed")

    @staticmethod
    def tax(s):
        return s * 0.1

    def __gt__(self, other):
        return self.__salary > other.__salary

e1 = Employee(10000)
e2 = Employee(12000)

e1.set_salary(11000)

print(Employee.tax(10000))
print(e2 > e1)
print(e1.history)
```

#### 30. Design a complete system combining:
#### • Abstract class Vehicle
#### • Multiple inheritance (Electric + Fuel)
#### • Operator overloading (compare by speed)
#### • Private attributes + getter/setter
#### • Custom exceptions
#### • File handling to save vehicle details
#### • Duck Typing for start() method

```python
from abc import ABC, abstractmethod

class SpeedError(Exception):
    pass

class Vehicle(ABC):
    def __init__(self, speed):
        self.__speed = speed

    def get_speed(self):
        return self.__speed

    def set_speed(self, s):
        if s < 0:
            raise SpeedError("Invalid Speed")
        self.__speed = s

    @abstractmethod
    def start(self):
        pass

    def __gt__(self, other):
        return self.__speed > other.__speed

class Electric:
    def charge(self):
        print("Charging")

class Fuel:
    def fill(self):
        print("Fuel Filled")

class Car(Vehicle, Electric, Fuel):
    def __init__(self, speed):
        super().__init__(speed)

    def start(self):
        print("Car Started")

c1 = Car(100)
c2 = Car(80)

print(c1 > c2)

f = open("vehicle.txt","w")
f.write(str(c1.get_speed()))
f.close()
```
