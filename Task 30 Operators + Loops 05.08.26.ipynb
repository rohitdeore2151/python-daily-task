#### 1. Take two numbers from user. Perform all arithmetic operations and also check which number is greater using comparison operators.

```python
a = int(input("Enter First Number: "))
b = int(input("Enter Second Number: "))

print("Addition =", a + b)
print("Substraction =",a - b)
print("Multiplication =",a * b)
print("Division =",a / b)
print("Modulus =",a // b)
print("Substraction =",a % b)
print("Equal =",a == b)
print("Not Equal =",a != b)
print("Greater Than =",a > b)
print("Less Than=",a < b)
print("Logical AND =",a > 0 and b > 0)
print("Logical OR=",a > 0 or b > 0)
print("Logical NOT=",not(a > b))
print(f"Nummber Is Greter :{max(a,b)}")
```

#### 2. Print all numbers from 1 to 50. Skip multiples of 5 using continue and stop completely when number 40 is reached using break.

```python
for i in range(1, 51):
    if i == 40:
        break     

    if i % 5 == 0:
        continue  

    print(i, end=" ")
```

#### 3. Take numbers from user until they enter negative number. Calculate sum and average using while loop.

```python
sum = 0
count = 0

while True:
    num = int(input("Enter a number: "))

    if num < 0:
        break

    sum += num
    count += 1

if count > 0:
    average = sum / count
    print("Sum =", sum)
    print("Average =", average)
else:
    print("No positive numbers were entered.")
```

#### 5. Find the factorial of a number using for loop and also using while loop.

```python
num = int(input("Enter a number: "))

fact = 1

for i in range(1, num + 1):
    fact *= i

print("Factorial =", fact)
```

#### 6. Take a string and perform: convert to uppercase, count occurrences of ‘a’, replace spaces with -, and check if it starts with “Hello”.

```python
text = input("Enter a string: ")

print("Uppercase:", text.upper())

print("Count of 'a':", text.count('a'))

print("Replace spaces:", text.replace(" ", "-"))

if text.startswith("Hello"):
    print("The string starts with 'Hello'.")
else:
    print("The string does not start with 'Hello'.")
```

#### 7. Create a list of 6 numbers. Use append, insert, remove, pop, sort (descending) and reverse.

```python
lists =  [1,2,3,4,5,6]


print("Original :",lists)

lists.append(10)
print("After append:", lists)
lists.insert(2, 15)
print("After insert:", lists)
lists.remove(5)
print("After remove:", lists)
lists.pop()
print("After pop:", lists)
lists.sort(reverse=True)
print("After sorting (descending):", lists)
lists.reverse()
print("After reverse:", lists)
```

#### 8. Create a dictionary of 4 students (name:marks). Print only those students who scored more than 80 using items().

```python
students = {
    "Rohit": 85,
    "yash": 78,
    "Gokul": 92,
    "Radha": 81
}

print("Students who scored more than 80:")

for name, marks in students.items():
    if marks > 80:
        print(name, ":", marks)
```

#### 9. Create two sets and perform union, intersection, difference. Also add and discard  elements.

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

print("Set 1:", set1)
print("Set 2:", set2)

print("Union:", set1.union(set2))

print("Intersection:", set1.intersection(set2))

print("Difference (set1 - set2):", set1.difference(set2))

set1.add(9)
print("After adding 9 to set1:", set1)

set2.discard(8)
print("After discarding 8 from set2:", set2)
```

#### OOP + Encapsulation (10-14)

### 10. Create a class Student with private attribute __marks. Add getter and setter with validation (0–100).

```python
class Student:
    def __init__(self):
        self.__marks = 0   

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks! Marks should be between 0 and 100.")

    def get_marks(self):
        return self.__marks


s = Student()

s.set_marks(85)
print("Marks:", s.get_marks())

s.set_marks(120)
print("Marks:", s.get_marks())
```
