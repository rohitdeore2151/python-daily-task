```python
#Question 7:
#Python
# Take a word from user.
# Check if the word is a palindrome or not.
# (Example: "madam", "radar")
```

```python
word = "yash"
if word == word[::-1]:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
#Question 8:
#Python
# Take email as input (e.g: student@gmail.com)
# Print only the username (part before @)
# Print only the domain (part after @)
```

## Bonus Mixed Question (String + Logical)

```python
#Python
# Take a password from user.
# Check if password is strong:
# - Length >= 8
# - Contains '@' or '#'
# Print "Strong Password" or "Weak Password"
```

```python
password = input("Enter the password")
if len(password) >= 8 and ('@' in password or '#' in password):
    print("Strong Password")
else:
    print("Weak Password")
```

```python
#1. Logical Operators
#Python
# Take three subject marks from user (English, Math, Science).
# Check if the student has passed:
# All marks should be >= 40 (use 'and' operator)
# If any one subject is < 40, print "Fail"
```

```python
# Take marks from the user
english = int(input("Enter English marks: "))
math = int(input("Enter Math marks: "))
science = int(input("Enter Science marks: "))

# Check if the student has passed
if english >= 40 and math >= 40 and science >= 40:
    print("Pass")
else:
    print("Fail")
```
