## Basic File Handling (1-10)

#### 1. Write a program to create a text file and write your name and age into it.

```python
with open("info.txt", "w") as file:
    file.write("Name: Rohit\n")
    file.write("Age: 20")
print("File created successfully.")
```

#### 2. Read the content of a text file and print it on the screen.

```python
with open("info.txt", "r") as file:
    print(file.read())
```

#### 3. Write a program to append some text to an existing file.

```python
with open("info.txt", "a") as file:
    file.write("\nCity: Hyderabad")
print("Text appended successfully.")
```

#### 4. Count the number of lines in a text file.

```python
with open("info.txt", "r") as file:
    lines = file.readlines()

print("Number of lines:", len(lines))
```

#### 5. Count the number of words in a text file.

```python
with open("info.txt", "r") as file:
    text = file.read()

words = text.split()
print("Number of words:", len(words))
```

#### 6. Count the number of characters in a text file.

```python
with open("info.txt", "r") as file:
    text = file.read()

print("Number of characters:", len(text))
```

#### 7. Read a file line by line using a loop.

```python
with open("info.txt", "r") as file:
    for line in file:
        print(line.strip())
```

#### 8. Write a program to copy content from one file to another file.

```python
with open("info.txt", "r") as source:
    data = source.read()

with open("copy.txt", "w") as destination:
    destination.write(data)

print("File copied successfully.")
```

#### 9. Check if a file exists or not before reading it.

```python
import os

if os.path.exists("info.txt"):
    with open("info.txt", "r") as file:
        print(file.read())
else:
    print("File does not exist.")
```

#### 10. Write a program to delete a file (use os module).

```python
import os

if os.path.exists("copy.txt"):
    os.remove("copy.txt")
    print("File deleted.")
else:
    print("File not found.")
```

#### 15. Read a file and display the longest word in it.

```python
with open("sample.txt", "r") as file:
    words = file.read().split()

longest = max(words, key=len)

print("Longest word:", longest)
```

#### 16. Write a program to merge two text files into a third file.

```python
with open("file1.txt", "r") as f1:
    data1 = f1.read()

with open("file2.txt", "r") as f2:
    data2 = f2.read()

with open("merged.txt", "w") as f3:
    f3.write(data1)
    f3.write("\n")
    f3.write(data2)

print("Files merged successfully.")
```

#### 17. Store a list of numbers in a file, then read and calculate their sum.


```python
numbers = [10, 20, 30, 40, 50]

with open("numbers.txt", "w") as file:
    for num in numbers:
        file.write(str(num) + "\n")

total = 0

with open("numbers.txt", "r") as file:
    for line in file:
        total += int(line)

print("Sum =", total)
```

#### 18. Create a file and write 10 lines. Then read only the first 5 lines.

```python
with open("lines.txt", "w") as file:
    for i in range(1, 11):
        file.write(f"Line {i}\n")

with open("lines.txt", "r") as file:
    for i in range(5):
        print(file.readline().strip())
```

#### 19. Write a program to replace a word in a file with another word.

```python
old = input("Enter word to replace: ")
new = input("Enter new word: ")

with open("sample.txt", "r") as file:
    text = file.read()

text = text.replace(old, new)

with open("sample.txt", "w") as file:
    file.write(text)

print("Word replaced successfully.")
```

#### 20. Read a CSV-like data from a text file and print it in tabular form.

```python
with open("data.txt", "r") as file:
    for line in file:
        print(*line.strip().split(","), sep="\t")
```

#### 22. Write a program that reads a log file and counts the number of “ERROR” messages.

```python
count = 0

with open("log.txt", "r") as file:
    for line in file:
        if "ERROR" in line:
            count += 1

print("Number of ERROR messages:", count)
```

#### 23. Take multiple lines of text from user until they type “STOP”, then save all lines in a file.

```python
with open("notes.txt", "w") as file:
    while True:
        line = input("Enter text: ")

        if line == "STOP":
            break

        file.write(line + "\n")

print("Data Saved Successfully.")
```

#### 24. Write a program to encrypt the content of a file (simple character shifting) and save it in another file.

```python
shift = 3

with open("sample.txt", "r") as file:
    text = file.read()

encrypted = ""

for ch in text:
    encrypted += chr(ord(ch) + shift)

with open("encrypted.txt", "w") as file:
    file.write(encrypted)

print("Encryption Completed.")
```

#### 25. Read a file containing student marks, calculate average, and write the result in another file.

```python
with open("marks.txt", "r") as file:
    marks = [int(line) for line in file]

average = sum(marks) / len(marks)

with open("average.txt", "w") as file:
    file.write("Average Marks = " + str(average))

print("Average Calculated.")
```

#### 26. Create a program that keeps a record of daily expenses in a file and can show total expenses.

```python
expense = float(input("Enter today's expense: "))

with open("expenses.txt", "a") as file:
    file.write(str(expense) + "\n")

total = 0

with open("expenses.txt", "r") as file:
    for line in file:
        total += float(line)

print("Total Expenses =", total)
```

#### 27. Write a program to remove all blank lines from a file.

```python
with open("sample.txt", "r") as file:
    lines = file.readlines()

with open("clean.txt", "w") as file:
    for line in lines:
        if line.strip():
            file.write(line)

print("Blank Lines Removed.")
```

#### 28. Read a file and write only the unique lines into another file (remove duplicates).

```python
seen = set()

with open("sample.txt", "r") as file, open("unique.txt", "w") as out:
    for line in file:
        if line not in seen:
            seen.add(line)
            out.write(line)

print("Duplicate Lines Removed.")
```

#### 29. Create a menu-driven program:
##### • Write to file
##### • Read file
##### • Append to file
##### • Exit

```python
while True:
    print("\n1. Write")
    print("2. Read")
    print("3. Append")
    print("4. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        text = input("Enter Text: ")
        with open("menu.txt", "w") as file:
            file.write(text)

    elif choice == "2":
        try:
            with open("menu.txt", "r") as file:
                print(file.read())
        except FileNotFoundError:
            print("File does not exist.")

    elif choice == "3":
        text = input("Enter Text: ")
        with open("menu.txt", "a") as file:
            file.write("\n" + text)

    elif choice == "4":
        print("Program Closed.")
        break

    else:
        print("Invalid Choice.")
```

#### 30. Write a program that handles file exceptions (FileNotFoundError, PermissionError) properly while reading and writing files.

```python
try:
    with open("input.txt", "r") as file:
        data = file.read()

    with open("output.txt", "w") as file:
        file.write(data)

except FileNotFoundError:
    print("Error: File not found.")

except PermissionError:
    print("Error: Permission denied.")
```
