#### 1. Create a class Student with instance attributes name and age. Create two objects and print their details.

```python
class Student :

    def __init__(self,name,age):
        self.Student_Name = name
        self.Student_Age = age

    def student_details(self):
        print(f"Student Name :{self.Student_Name}")
        print(f"Student Age : {self.Student_Age}")

std1 = Student("Rohit",22)
std2 = Student("Gokul",23)
std1.student_details()
std2.student_details()
```

#### 4. Create class Employee with attributes emp_id, name, salary. Create 3 objects.

#### 5. Write a class Mobile with attributes brand, price, color. Print object using __str__method.

```python
class Mobile:

    def __init__(self, brand, price, color):
        self.brand = brand
        self.price = price
        self.color = color

    def __str__(self):
        return f"Brand: {self.brand}, Price: {self.price}, Color: {self.color}"


m1 = Mobile("Samsung", 25000, "Black")
m2 = Mobile("Apple", 75000, "White")
m3 = Mobile("OnePlus", 35000, "Blue")

# Printing objects
print(m1)
print(m2)
print(m3)
```

#### 6. Create class Rectangle with attributes length and width. Add method to calculate area.

```python
class Rectangle:

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        area = self.length * self.width
        return area

r1 = Rectangle(10, 5)

print("Length:", r1.length)
print("Width:", r1.width)
print("Area of Rectangle:", r1.calculate_area())
```

#### 13. In class Circle, add class attribute pi = 3.14. Use it in area calculation method.

```python
class Circle:
    pi = 3.14  

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return Circle.pi * self.radius * self.radius


c1 = Circle(5)

print("Area of Circle:", c1.area())
```

#### 19. In class Product, make __price private. Try accessing from outside.

#### 20. Create class User with private __password. Demonstrate name mangling.

```python
class User:
    def __init__(self, username, password):
        self.username = username
        self.__password = password   # private variable


# Create object
user = User("Rahul", "12345")

# Access using name mangling
print(user._User__password)
```
