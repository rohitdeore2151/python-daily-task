#### 2. Plot two lines representing Product A and Product B sales for 6 months. Add a legend and different line styles.

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
product_A = [120, 150, 170, 160, 180, 200]
product_B = [100, 130, 140, 150, 170, 190]

# Plot lines
plt.plot(months, product_A, marker='o', linestyle='-', label="Product A")
plt.plot(months, product_B, marker='<', linestyle='--', label="Product B")

# Title and labels
plt.title("Product A and Product B Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

# Legend and grid
plt.legend()
plt.grid(True)

# Show plot
plt.show()
```

#### 3. Create a bar chart showing the revenue of 5 different products. Display the revenue value on top of each bar.

```python
products = ["Product A", "Product B", "Product C", "Product D", "Product E"]
revenue = [50000, 70000, 60000, 80000, 65000]

bars = plt.bar(products, revenue)

for bar in bars:
    plt.text(bar.get_x() + bar.get_width()/2,
             bar.get_height(),
             str(bar.get_height()),
             ha='center', va='bottom')

plt.title("Revenue of 5 Products")
plt.xlabel("Products")
plt.ylabel("Revenue")

plt.grid(axis='y')
plt.show()
```

#### 4. Create a horizontal bar chart showing the performance of 6 employees based on their sales.

```python
employees = ["Emp1", "Emp2", "Emp3", "Emp4", "Emp5", "Emp6"]
sales = [120, 150, 100, 180, 140, 160]

plt.figure(figsize=(8, 5))
bars = plt.barh(employees, sales)

for bar in bars:
    plt.text(bar.get_width() + 2,
             bar.get_y() + bar.get_height()/2,
             str(bar.get_width()),
             va='center')
    
plt.title("Employee Sales Performance")
plt.xlabel("Sales")
plt.ylabel("Employees")

plt.grid(axis='x')
plt.show()
```

#### 11.Create a monthly sales chart and identify the maximum-sales month.
#### Highlight that point using annotate().

```python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
sales = [20,25,30,28,35,40,38,45,50,48,55,60]

plt.figure(figsize=(10,5))
plt.plot(months,sales,marker='o')

max_sale = max(sales)
max_month = months[sales.index(max_sale)]

plt.annotate("Highest Sales",
             xy=(max_month,max_sale),
             xytext=(max_month,max_sale+5),
             arrowprops=dict(arrowstyle="->"))

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```
