### Data Type & Invalid Values

#### 11.The Age column contains numbers as well as values like "unknown" and "twenty five". Clean this column and convert it into a proper numeric column.

```python
import numpy as np
import pandas as pd

df['age'] = df['age'].replace({
    'unknown': np.nan,
    'twenty five': 25
})

df['age'] = pd.to_numeric(df['age'], errors='coerce')

print(df['age'])
print(df['age'].dtype)
```

#### 12.The Annual Income column contains values such as 650000 INR, ₹720000, 1.2L, and N/A. Clean the column and convert the valid values into numeric format.

```python
import numpy as np
import pandas as pd

df['annual_income'] = (
    df['annual_income']
      .astype(str)
      .str.replace('INR', '', regex=False)   # Remove INR
      .str.replace('₹', '', regex=False)     # Remove ₹
      .str.strip()
)

df['annual_income'] = df['annual_income'].replace(
    r'(\d+\.?\d*)L',
    lambda x: str(float(x.group(1)) * 100000),
    regex=True
)

df['annual_income'] = df['annual_income'].replace('N/A', np.nan)

df['annual_income'] = pd.to_numeric(df['annual_income'], errors='coerce')

print(df['annual_income'])
print(df['annual_income'].dtype)
```
