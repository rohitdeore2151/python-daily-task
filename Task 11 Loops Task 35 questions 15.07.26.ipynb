# Basic For Loop (1-8)

```python
# 1. Print numbers from 1 to 10 using for loop.

for i in range (1,11):
    print(i)
```

```python
# 2. Take a number n from user and print sum of first n natural numbers using for loop.

n = int(input("Enter a number: "))

total = 0

for i in range(1, n + 1):
    total += i

print("Sum of first", n, "natural numbers is:", total)
```

```python
# 3. Print multiplication table of a given number using for loop.

num = int(input("Enter The Number"))

for i in range(1,11):
   print(num, "x", i, "=", num * i)
```

```python
# 4. Print all even numbers from 1 to 20 using for loop.

for i in range(1,20):
    if i % 2 == 0:
        print("Even Number",i)
    else:
        print("Odd Number",i)
        
```

```python
# 5. Count how many numbers between 1 to 50 are divisible by 3 using for loop.

count = 0
for i in range (1,51):
    if i % 3 == 0:
        print(i)
        count +=1
print("total number divisible by 3:",count)
   
```

```python
# 6. Print characters of a string one by one using for loop.

character = input("Enter The Any String")

for ch in character:
    print(ch)
```

```python
# 7. Print numbers from 50 to 30 in descending order using for loop.

for i in range(50,29,-1):
    print(i)
```

```python
# 8. Find factorial of a number using for loop.

num = int(input("Enter The Number"))

factorial = 1

for i in range(1, num +1):
    factorial = factorial * i
print("factorial Number", num, "is", factorial)
```

```python
# 9. Print numbers from 1 to 15 using while loop.

i = 1
while i <= 15:
    print(i)
    i += 1
    
```

```python
# 10. Keep taking numbers from user until they enter 0, then print sum.

sum =  0 

num = int(input("Enter The Number:"))

while num != 0:
    sum += num
    num = int(input("Enter The Number:"))
    print("sum of the number is:",sum)
```

```python
# 11. Reverse a number using while loop.

num = int(input("Enter The Number"))

rev = 0

digit = 0

while num >0:
     digit = num % 10
     rev = rev * 10 + digit
     num = num // 10
print(rev)
    
```

```python
# 12. Check if a number is prime using while loop.

num = int(input("Enter The Number"))

if num <= 1:
    print(num,"is not a prime number.")
else:
    i = 2
    is_prime = True
   
    while i < num:
        if num % i == 0:
            is_prime = False
            break
            i += 1
        if is_prime:
            print(num,"is a prime number.")
        else:
            print(num,"is not a prime number.")
    
```

```python
# 13. Print Fibonacci series up to n terms using while loop.

a =  0
b = 1
count = 1

while count <= 10:
    print(a, end="")
    c=a+b
    a=b
    b=c
    count +=1
```

```python
# 19. Print prime numbers between 1 to 100 using nested for loop.

for num in range(1, 101):
    is_prime = True 

    for i in range(2,num):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num)
```

```python
# 22. Print a hollow square pattern of size n using nested loop.

n = int(input("Enter the size of square: "))

for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# 23. Print Floyd’s Triangle using nested for loop.

n = int(input("Enter the number of rows: "))

num = 1

for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(num, end=" ")
        num += 1
    print()
```

# Break & Continue (24-35)

```python
# 24. Print numbers from 1 to 20, but stop when number 13 appears (use break).

for i in range(1, 21):
    if i == 13:
        break
    print(i)
```

```python
# 25. Print all numbers from 1 to 30 except multiples of 5 (use continue).

for i in range(1, 31):
    if i % 5 == 0:
        continue
    print(i)
```

```python
# 26. Take 5 numbers from user. If negative number entered, stop using break.

for i in range(5):
    num = int(input("Enter a number: "))

    if num < 0:
        print("Negative number entered. Stopping...")
        break

    print("You entered:", num)
```

```python
# 27. Print even numbers from 1 to 20 using continue.

for i in range(1,20):
    if i % 2 != 0:
        print("Even Number")
        continue
    print(i)
        
```

```python
# 28. Find first number divisible by both 3 and 7 between 1 to 100 using break.

for i in range(1,101):
    if i % 3 == 0 and i % 7 == 0:
         print("first number divisible by both 3 and 7 is :",i)
         break
   
```

```python
# 29. Take numbers until sum exceeds 100, use break.

total = 0

while True:
    num = int(input("Enter a number: "))
    total += num

    if total > 100:
        print("Sum exceeded 100.")
        break

print("Final Sum =", total)
```
