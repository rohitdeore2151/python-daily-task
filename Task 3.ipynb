## Here are 20 Practical Questions on Strings (Covering: Slicing, upper, lower, title, capitalize, isalpha, isdigit, isalnum, islower, find, index)

## String Slicing Questions

#### 1. Take a string from user and print the first 5 characte



```python
s = input("Enter a string: ")
print(s[:5])
```

#### 2. Print the last 4 characters using negative slicing

```python
s = input("Enter a string: ")
print(s[-4:])
```

#### 3. Print characters from index 2 to 8

```python
s = input("Enter a string: ")
print(s[2:8])
```

#### 4. Print every 2nd character

```python
s = input("Enter a string: ")
print(s[::2])
```

#### 5. Print first name and last name separately

```python
name = input("Enter full name: ")

space = name.find(" ")

first_name = name[:space]
last_name = name[space+1:]

print("First Name:", first_name)
print("Last Name:", last_name)
```

#### 6. Case Conversion

```python
s = input("Enter a string: ")

print("Uppercase:", s.upper())
print("Lowercase:", s.lower())
print("Title Case:", s.title())
print("Capitalized:", s.capitalize())
```

#### 7. Convert sentence to title case

```python
sentence = input("Enter a sentence: ")

print("Original:", sentence)
print("Title Case:", sentence.title())
```

#### 8. Check whether lower() and upper() change the string permanently

```python
s = input("Enter a string: ")

print("Original:", s)
print("Lower:", s.lower())
print("Upper:", s.upper())

print("After methods:", s)
```

#### 9. Check if string contains only alphabets

```python
s = input("Enter a string: ")

print(s.isalpha())
```

#### 10. Check if input contains only digits

```python
num = input("Enter input: ")

print(num.isdigit())
```

#### 11. Check if string is alphanumeric

```python
s = input("Enter a string: ")

print(s.isalnum())
```

#### 12. Check if string is completely lowercase

```python
s = input("Enter a string: ")

print(s.islower())
```

#### 13. Password Validation

```python
password = input("Enter password: ")

if password.isalnum() and len(password) >= 8:
    print("Valid Password")
else:
    print("Invalid Password")
```

#### 14. Find first occurrence of a character

```python
s = input("Enter a string: ")
ch = input("Enter a character: ")

print(s.find(ch))
```

#### 15. Find a word using find()

```python
s = input("Enter a string: ")
word = input("Enter a word: ")

print(s.find(word))
```

#### 16. Find position of "Python" using index()

```python
sentence = input("Enter a sentence containing Python: ")

print(sentence.index("Python"))
```

#### 17. Difference between find() and index()

```python
s = input("Enter a string: ")

print("Using find():", s.find("z"))

print("Using index():")
print(s.index("z"))
```

#### 18. Full Name Operations

```python
name = input("Enter full name: ")

print("Uppercase:", name.upper())

space = name.find(" ")

print("First Space Position:", space)

print("First Name:", name[:space])
print("Last Name:", name[space+1:])
```

#### 19. Mobile Number Validation

```python
mobile = input("Enter mobile number: ")

if mobile.isdigit() and len(mobile) == 10:
    print("Valid Mobile Number")
else:
    print("Invalid Mobile Number")
```

#### 20. Reverse, Title Case & Palindrome Check

```python
s = input("Enter a string: ")

reverse = s[::-1]

print("Reverse:", reverse)

print("Title Case:", s.title())

if s == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```
